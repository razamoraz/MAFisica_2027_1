## 1. El Problema General de Sturm–Liouville

En un intervalo $x \in [a, b]$, la **forma canónica o autoadjunta** de la ecuación diferencial de Sturm–Liouville es:

$$\boxed{\mathcal{L}[y](x) \equiv -\frac{d}{dx}\left[p(x)\frac{dy}{dx}\right] + q(x)y(x) = \lambda w(x)y(x)}$$

o bien, expandiendo la derivada exterior:

$$\left[p(x)y'\right]' + \big(\lambda w(x) - q(x)\big)y = 0$$

### 1.1 Elementos del problema
- $p(x) > 0$ en $(a, b)$: función de rigidez o coeficiente diferencial principal (continuamente diferenciable, $C^1$).
- $w(x) > 0$ en $(a, b)$: **función de peso** (*weight function*), continua y estrictamente positiva casi en todas partes.
- $q(x)$: función potencial real y continua en $[a, b]$.
- $\lambda$: parámetro escalar de **autovalor** (*eigenvalue*).
- $y(x)$: **autofunción** (*eigenfunction*) asociada al autovalor $\lambda$.

### 1.2 Condiciones de frontera regulares (de tipo Sturm–Liouville)
Para garantizar que el operador diferencial $\mathcal{L}$ sea **formalmente autoadjunto** (simétrico respecto al producto interno en $L^2$), se imponen condiciones lineales homogéneas en los extremos:

$$\alpha_1 y(a) + \alpha_2 p(a) y'(a) = 0, \qquad \beta_1 y(b) + \beta_2 p(b) y'(b) = 0$$

donde $(\alpha_1, \alpha_2) \neq (0,0)$ y $(\beta_1, \beta_2) \neq (0,0)$.

### 1.3 Teoremas fundamentales de Sturm–Liouville
Para un problema de Sturm–Liouville regular:
1. **Realidad del espectro:** Todos los autovalores $\lambda_n$ son números reales.
2. **Espectro discreto y ordenado:** Existe una cantidad infinita contable de autovalores acotados inferiormente:
   $$\lambda_0 < \lambda_1 < \lambda_2 < \dots < \lambda_n < \dots \to +\infty$$
3. **Teorema de oscilación:** La autofunción $y_n(x)$ asociada a $\lambda_n$ tiene exactamente $n$ ceros simples en el intervalo abierto $(a, b)$.
4. **Ortogonalidad ponderada:** Si $\lambda_n \neq \lambda_m$, las autofunciones $y_n(x)$ e $y_m(x)$ son ortogonales con respecto al peso $w(x)$:
   $$\langle y_n, y_m \rangle_w \equiv \int_a^b y_n(x)\,y_m(x)\,w(x)\,dx = \delta_{nm}\,N_n$$
   donde $\delta_{nm}$ es la delta de Kronecker y $N_n = \|y_n\|_w^2$ es la norma cuadrática ponderada.
5. **Completitud:** Las autofunciones forman una **base ortogonal completa** del espacio de Hilbert ponderado $L^2_w([a,b])$.

## 2. Fourier: El Caso Prototípico Fundamental

Una de las conexiones conceptuales más profundas de la física matemática es entender que **el análisis de Fourier no es un tema aislado**, sino **el caso regular más elemental posible de Sturm–Liouville**:

$$p(x) = 1, \qquad q(x) = 0, \qquad w(x) = 1$$

Sustituyendo estos coeficientes en la ecuación general obtenemos la ecuación del oscilador armónico espacial:

$$\mathcal{L}[y] = -\frac{d}{dx}\left[1\cdot \frac{dy}{dx}\right] + 0 = -y'' = \lambda y \quad \Longrightarrow \quad y''(x) + \lambda y(x) = 0$$

### ¿De dónde salen los senos, los cosenos o ambos?
La ecuación diferencial es idéntica en todos los casos; **lo que determina la base de autofunciones son las condiciones de frontera** impuestas por la física del problema:

| Condición de frontera | Dominio | Autofunciones $y_n(x)$ | Autovalores $\lambda_n$ | Interpretación en Física |
|:---|:---:|:---|:---:|:---|
| **Dirichlet** (Frontera fija) | $[0, L]$ | $\sin\left(\frac{n\pi x}{L}\right), \quad n=1,2,\dots$ | $\left(\frac{n\pi}{L}\right)^2$ | **Partícula en una caja cuántica 1D** ($V=\infty$ en bordes); cuerda de guitarra con extremos fijos. |
| **Neumann** (Derivada nula) | $[0, L]$ | $\cos\left(\frac{n\pi x}{L}\right), \quad n=0,1,\dots$ | $\left(\frac{n\pi}{L}\right)^2$ | Barra con extremos perfectamente aislados térmicamente (flujo de calor $\partial T/\partial x = 0$); extremos libres. |
| **Periódicas** | $[-L, L]$ | $\left\{1, \cos\frac{n\pi x}{L}, \sin\frac{n\pi x}{L}\right\}$ o $\{e^{in\pi x/L}\}$ | $\left(\frac{n\pi}{L}\right)^2$ | Anillo unidimensional cerrado; electrón en un potencial periódico de celda cristalina (teorema de Bloch). |

> 💡 **Conclusión clave:** La partícula en una caja que estudiaste en física cuántica o la cuerda clásica fija **es** exactamente el problema de Sturm–Liouville con peso unitario y condiciones Dirichlet. Cuando decimos "desarrollo en serie de Fourier", simplemente estamos aplicando el **teorema de completez de Sturm–Liouville** para el operador $-d^2/dx^2$.

## 3. 📋 La Gran Tabla de Sturm–Liouville y Funciones Especiales

Cuando el dominio no es plano o uniforme (por ejemplo en geometrías esféricas, cilíndricas o potenciales cuánticos gaussianos), los coeficientes métricos transforman $p(x)$, $q(x)$ y el peso $w(x)$. A continuación se presenta la tabla canónica que unifica todas las grandes familias ortogonales de la física:

| Nombre | Dominio $[a,b]$ | Peso $w(x)$ | Primeras funciones ortogonales | Forma Sturm–Liouville |
|:---|:---:|:---:|:---|:---|
| **Fourier (completa, real)** | $[-L, L]$ | $1$ | $1,\;\cos\frac{\pi x}{L},\;\sin\frac{\pi x}{L},\;\cos\frac{2\pi x}{L},\;\sin\frac{2\pi x}{L},\dots$ | $-y'' = \lambda y,\quad \lambda_n=\left(\frac{n\pi}{L}\right)^2$ |
| **Fourier (compleja)** | $[-L, L]$ | $1$ | $e^{in\pi x/L},\; n\in\mathbb{Z}$ | $-y'' = \lambda y,\quad \lambda_n=\left(\frac{n\pi}{L}\right)^2$ |
| **Fourier seno (half-range)** | $[0, L]$ | $1$ | $\sin\frac{\pi x}{L},\;\sin\frac{2\pi x}{L},\;\sin\frac{3\pi x}{L},\dots$ | $-y'' = \lambda y,\; y(0)=y(L)=0$ |
| **Fourier coseno (half-range)** | $[0, L]$ | $1$ | $1,\;\cos\frac{\pi x}{L},\;\cos\frac{2\pi x}{L},\dots$ | $-y'' = \lambda y,\; y'(0)=y'(L)=0$ |
| **Legendre** $P_n(x)$ | $[-1, 1]$ | $1$ | $P_0=1,\; P_1=x,\; P_2=\frac{1}{2}(3x^2-1),\; P_3=\frac{1}{2}(5x^3-3x)$ | $\left[(1-x^2)P_n'\right]' + n(n+1)P_n = 0$ |
| **Chebyshev I** $T_n(x)$ | $[-1, 1]$ | $(1-x^2)^{-1/2}$ | $T_0=1,\; T_1=x,\; T_2=2x^2-1,\; T_3=4x^3-3x$ | $\left[(1-x^2)^{1/2}T_n'\right]' + \frac{n^2}{(1-x^2)^{1/2}}T_n = 0$ |
| **Chebyshev II** $U_n(x)$ | $[-1, 1]$ | $(1-x^2)^{1/2}$ | $U_0=1,\; U_1=2x,\; U_2=4x^2-1,\; U_3=8x^3-4x$ | $\left[(1-x^2)^{1/2}U_n'\right]' + \frac{n(n+2)}{(1-x^2)^{1/2}}U_n = 0$ |
| **Hermite** $H_n(x)$ | $(-\infty, \infty)$ | $e^{-x^2}$ | $H_0=1,\; H_1=2x,\; H_2=4x^2-2,\; H_3=8x^3-12x$ | $\left[e^{-x^2}H_n'\right]' + 2n e^{-x^2}H_n = 0$ |
| **Laguerre** $L_n(x)$ | $[0, \infty)$ | $e^{-x}$ | $L_0=1,\; L_1=1-x,\; L_2=1-2x+\frac{1}{2}x^2,\; L_3=1-3x+\frac{3}{2}x^2-\frac{1}{6}x^3$ | $\left[x e^{-x}L_n'\right]' + n e^{-x}L_n = 0$ |
| **Laguerre generalizado** $L_n^{(\alpha)}(x)$ | $[0, \infty)$ | $x^\alpha e^{-x},\; \alpha>-1$ | $L_0^{(\alpha)}=1,\; L_1^{(\alpha)}=1+\alpha-x,\; L_2^{(\alpha)}=\frac{1}{2}(x^2-2(\alpha+2)x+(\alpha+1)(\alpha+2))$ | $\left[x^{\alpha+1}e^{-x}(L_n^{(\alpha)})'\right]' + n x^\alpha e^{-x}L_n^{(\alpha)} = 0$ |
| **Jacobi** $P_n^{(\alpha,\beta)}(x)$ | $[-1, 1]$ | $(1-x)^\alpha(1+x)^\beta$ | $P_0^{(\alpha,\beta)}=1,\; P_1^{(\alpha,\beta)}=\frac{1}{2}(\alpha+\beta+2)x+\frac{1}{2}(\alpha-\beta)$ | $\left[(1-x^2)w(P_n^{(\alpha,\beta)})'\right]' + n(n+\alpha+\beta+1)wP_n^{(\alpha,\beta)} = 0$ |
| **Gegenbauer** $C_n^{(\alpha)}(x)$ | $[-1, 1]$ | $(1-x^2)^{\alpha-1/2},\; \alpha>-1/2$ | $C_0^{(\alpha)}=1,\; C_1^{(\alpha)}=2\alpha x,\; C_2^{(\alpha)}=2\alpha(1+\alpha)x^2-\alpha$ | $\left[(1-x^2)^{\alpha+1/2}(C_n^{(\alpha)})'\right]' + n(n+2\alpha)w C_n^{(\alpha)} = 0$ |
| **Bessel** $J_\nu(k_n x)$ | $[0, a]$ | $x$ | $J_0(k_1 x),\; J_1(k_1 x),\; J_0(k_2 x),\; J_1(k_2 x), \dots$ | $\left[x J_\nu'\right]' + \left(k^2 x - \frac{\nu^2}{x}\right)J_\nu = 0$ |

> 🔍 **Observación sobre problemas singulares:** Cuando el intervalo es infinito (como en Hermite o Laguerre) o cuando $p(x)=0$ en algún extremo (como en Legendre donde $p(\pm 1)=1-(\pm 1)^2=0$), el problema se denomina **singular**. En estos casos, las condiciones de frontera regulares se sustituyen por la exigencia física de que la solución sea **regular y de cuadrado integrable** en los extremos.

## 4. 🧮 Motor Simbólico: Verificación Formal del Operador de Sturm–Liouville con `SymPy`

Antes de graficar numéricamente, utilicemos el motor de álgebra computacional simbólica (`SymPy`) para implementar directamente la definición formal del operador:

$$\mathcal{L}[y] = -\frac{d}{dx}\left[p(x)\,\frac{dy}{dx}\right] + q(x)\,y$$

y verificar de manera exacta que para cada familia de la física matemática, al aplicar $\mathcal{L}$ sobre la autofunción $y(x)$ se satisface idénticamente:

$$\mathcal{L}[y](x) - \lambda\,w(x)\,y(x) \equiv 0$$

```python
# ============================================================
# 🔬 Verificación Simbólica del Operador de Sturm–Liouville
# ============================================================

def operador_SL(y_expr, p_expr, q_expr, var=x):
    """
    Aplica formalmente el operador diferencial de Sturm–Liouville:
        L[y] = -d/dx [ p(x) * dy/dx ] + q(x) * y
    utilizando cálculo diferencial simbólico exacto con SymPy.
    """
    derivada_interna = smp.diff(y_expr, var)
    termino_flujo = p_expr * derivada_interna
    termino_dif = -smp.diff(termino_flujo, var)
    return termino_dif + q_expr * y_expr

print("=" * 75)
print("  COMPROBACIÓN SIMBÓLICA CON SYMPY: L[y] - λ·w(x)·y == 0")
print("=" * 75)

# ------------------------------------------------------------
# 1. Caso Fourier Seno: p=1, q=0, w=1, λ_n = (n*pi/L)^2
# ------------------------------------------------------------
y_fourier = smp.sin(n * smp.pi * x / L)
L_fourier = operador_SL(y_fourier, p_expr=1, q_expr=0)
lambda_fourier = (n * smp.pi / L)**2
residuo_fourier = smp.simplify(L_fourier - lambda_fourier * 1 * y_fourier)

print(f"1. Fourier Seno sin(n·π·x/L):")
print(f"   L[y]   = {L_fourier}")
print(f"   λ·w·y  = {lambda_fourier * y_fourier}")
print(f"   Residuo simbólico = {residuo_fourier}  ✓ (¡Idénticamente nulo!)")
print("-" * 75)

# ------------------------------------------------------------
# 2. Caso Legendre P_n(x): p=1-x^2, q=0, w=1, λ_n = n(n+1)
# ------------------------------------------------------------
for orden in [1, 2, 3]:
    Pn = smp.legendre(orden, x)
    L_leg = operador_SL(Pn, p_expr=1 - x**2, q_expr=0)
    lambda_leg = orden * (orden + 1)
    residuo_leg = smp.simplify(L_leg - lambda_leg * 1 * Pn)
    print(f"2. Legendre P_{orden}(x) = {Pn}:")
    print(f"   L[P_{orden}] = {smp.factor(L_leg)}")
    print(f"   λ_{orden}·w·P_{orden} = {lambda_leg * Pn}")
    print(f"   Residuo simbólico = {residuo_leg}  ✓")
print("-" * 75)

# ------------------------------------------------------------
# 3. Caso Hermite H_n(x): p=exp(-x^2), q=0, w=exp(-x^2), λ_n = 2n
# ------------------------------------------------------------
for orden in [1, 2]:
    Hn = smp.hermite(orden, x)
    L_herm = operador_SL(Hn, p_expr=smp.exp(-x**2), q_expr=0)
    lambda_herm = 2 * orden
    w_herm = smp.exp(-x**2)
    residuo_herm = smp.simplify(L_herm - lambda_herm * w_herm * Hn)
    print(f"3. Hermite H_{orden}(x) = {Hn}:")
    print(f"   Residuo simbólico = {residuo_herm}  ✓")
print("=" * 75)
```

## 5. 📐 Generación Simbólica y Evaluación Numérica con `lambdify`

Una de las fortalezas clave de `SymPy` en física matemática es la función **`lambdify`**:
1. Define la familia analíticamente como una expresión simbólica exacta (mostrando las fracciones de los polinomios).
2. Compila la expresión simbólica en una función `NumPy` optimizada en lenguaje C vectorizado para evaluar y graficar cientos de miles de puntos instantáneamente.

```python
# ============================================================
# 📊 Galería de Familias Ortogonales de Sturm–Liouville
# ============================================================

# Paleta armónica de colores de alto contraste
colores = ['#58a6ff', '#3fb950', '#d29922', '#f85149', '#bc8cff']

def evaluar_simb_vector(expr_sym, xx_pts):
    """Convierte una expresión SymPy a NumPy y garantiza salida vectorizada incluso para constantes."""
    fn = smp.lambdify(x, expr_sym, 'numpy')
    val = fn(xx_pts)
    return np.broadcast_to(val, xx_pts.shape)

fig, axes = plt.subplots(3, 3, figsize=(16, 13))
fig.patch.set_facecolor('#0d1117')

# ------------------------------------------------------------
# 1. Fourier Seno (Condiciones Dirichlet en [0, 1])
# ------------------------------------------------------------
ax = axes[0, 0]
xx = np.linspace(0, 1, 400)
for k in range(1, 5):
    expr_sym = smp.sin(k * smp.pi * x)
    ax.plot(xx, evaluar_simb_vector(expr_sym, xx), label=rf'${smp.latex(expr_sym)}$', color=colores[k-1], lw=2)
ax.set_title('1. Fourier Seno: -y\'\' = λ y,  y(0)=y(1)=0', fontweight='bold')
ax.set_xlabel('x'); ax.set_ylabel('y_n(x)')
ax.legend(fontsize=8, framealpha=0.25); ax.grid(True)

# ------------------------------------------------------------
# 2. Fourier Coseno (Condiciones Neumann en [0, 1])
# ------------------------------------------------------------
ax = axes[0, 1]
xx = np.linspace(0, 1, 400)
for k in range(0, 4):
    expr_sym = smp.cos(k * smp.pi * x)
    ax.plot(xx, evaluar_simb_vector(expr_sym, xx), label=rf'${smp.latex(expr_sym)}$', color=colores[k], lw=2)
ax.set_title('2. Fourier Coseno: -y\'\' = λ y,  y\'(0)=y\'(1)=0', fontweight='bold')
ax.set_xlabel('x'); ax.set_ylabel('y_n(x)')
ax.legend(fontsize=8, framealpha=0.25); ax.grid(True)

# ------------------------------------------------------------
# 3. Fourier Completa (Periódica en [-1, 1])
# ------------------------------------------------------------
ax = axes[0, 2]
xx = np.linspace(-1, 1, 400)
ax.plot(xx, np.cos(1*np.pi*xx), label=r'$\cos(\pi x)$', color='#58a6ff', lw=2)
ax.plot(xx, np.sin(1*np.pi*xx), label=r'$\sin(\pi x)$', color='#3fb950', lw=2, ls='--')
ax.plot(xx, np.cos(2*np.pi*xx), label=r'$\cos(2\pi x)$', color='#d29922', lw=2)
ax.plot(xx, np.sin(2*np.pi*xx), label=r'$\sin(2\pi x)$', color='#f85149', lw=2, ls='--')
ax.set_title('3. Fourier Completa: -y\'\' = λ y  (Periódico)', fontweight='bold')
ax.set_xlabel('x'); ax.set_ylabel('y_n(x)')
ax.legend(fontsize=8, framealpha=0.25); ax.grid(True)

# ------------------------------------------------------------
# 4. Polinomios de Legendre P_n(x) en [-1, 1] generados con SymPy
# ------------------------------------------------------------
ax = axes[1, 0]
xx = np.linspace(-1, 1, 400)
for k in range(4):
    expr_sym = smp.legendre(k, x)
    ax.plot(xx, evaluar_simb_vector(expr_sym, xx), label=rf'$P_{k}(x)={smp.latex(expr_sym)}$', color=colores[k], lw=2)
ax.set_title('4. Legendre: [(1-x²)P_n\']\' + n(n+1)P_n = 0', fontweight='bold')
ax.set_xlabel('x'); ax.set_ylabel('P_n(x)')
ax.legend(fontsize=8, framealpha=0.25); ax.grid(True)

# ------------------------------------------------------------
# 5. Chebyshev de 1ª especie T_n(x) en [-1, 1] generados con SymPy
# ------------------------------------------------------------
ax = axes[1, 1]
xx = np.linspace(-1, 1, 400)
for k in range(4):
    expr_sym = smp.chebyshevt(k, x)
    ax.plot(xx, evaluar_simb_vector(expr_sym, xx), label=rf'$T_{k}(x)={smp.latex(expr_sym)}$', color=colores[k], lw=2)
ax.set_title('5. Chebyshev I: w(x) = (1-x²)^(-1/2)', fontweight='bold')
ax.set_xlabel('x'); ax.set_ylabel('T_n(x)')
ax.legend(fontsize=8, framealpha=0.25); ax.grid(True)

# ------------------------------------------------------------
# 6. Polinomios de Hermite H_n(x) en (-∞, ∞) con SymPy
# ------------------------------------------------------------
ax = axes[1, 2]
xx = np.linspace(-2.5, 2.5, 400)
for k in range(4):
    expr_sym = smp.hermite(k, x)
    ax.plot(xx, evaluar_simb_vector(expr_sym, xx), label=rf'$H_{k}(x)={smp.latex(expr_sym)}$', color=colores[k], lw=2)
ax.set_title('6. Hermite: [exp(-x²)H_n\']\' + 2n exp(-x²)H_n = 0', fontweight='bold')
ax.set_xlabel('x'); ax.set_ylabel('H_n(x)')
ax.legend(fontsize=8, framealpha=0.25); ax.grid(True)

# ------------------------------------------------------------
# 7. Polinomios de Laguerre L_n(x) en [0, ∞) con SymPy
# ------------------------------------------------------------
ax = axes[2, 0]
xx = np.linspace(0, 8, 400)
for k in range(4):
    expr_sym = smp.laguerre(k, x)
    ax.plot(xx, evaluar_simb_vector(expr_sym, xx), label=rf'$L_{k}(x)={smp.latex(expr_sym)}$', color=colores[k], lw=2)
ax.set_title('7. Laguerre: [x exp(-x)L_n\']\' + n exp(-x)L_n = 0', fontweight='bold')
ax.set_xlabel('x'); ax.set_ylabel('L_n(x)')
ax.legend(fontsize=8, framealpha=0.25); ax.grid(True)

# ------------------------------------------------------------
# 8. Jacobi P_n^{(1,1)}(x) con SymPy
# ------------------------------------------------------------
ax = axes[2, 1]
xx = np.linspace(-1, 1, 400)
for k in range(1, 4):
    expr_sym = smp.jacobi(k, 1, 1, x)
    ax.plot(xx, evaluar_simb_vector(expr_sym, xx), label=rf'$P_{k}^{{(1,1)}}={smp.latex(expr_sym)}$', color=colores[k], lw=2)
ax.set_title('8. Jacobi: w(x) = (1-x)^α(1+x)^β', fontweight='bold')
ax.set_xlabel('x'); ax.set_ylabel('P_n^{(1,1)}(x)')
ax.legend(fontsize=8, framealpha=0.25); ax.grid(True)

# ------------------------------------------------------------
# 9. Funciones de Bessel J_0, J_1, J_2 en [0, 10]
# ------------------------------------------------------------
ax = axes[2, 2]
xx = np.linspace(0, 12, 400)
for m in range(3):
    yy = sp.jv(m, xx)
    ax.plot(xx, yy, label=rf'$J_{m}(x)$', color=colores[m], lw=2)
ax.set_title('9. Bessel: [x J_ν\']\' + (k²x - ν²/x)J_ν = 0', fontweight='bold')
ax.set_xlabel('x'); ax.set_ylabel('J_m(x)')
ax.legend(fontsize=8, framealpha=0.25); ax.grid(True)

plt.tight_layout()
plt.show()
```

## 6. ⚖️ El Papel Físico y Matemático de la Función de Peso $w(x)$

¿Por qué no todas las familias tienen peso uniforme $w(x)=1$?

En álgebra lineal y mecánica cuántica, la ortogonalidad estándar $\int f(x)g(x)\,dx = 0$ asume una medida de Lebesgue plana. Sin embargo, en física casi nunca trabajamos en espacios unidimensionales planos sin potencial:

1. **Cambio de coordenadas (Factores de escala o Jacobiano):**
   - En coordenadas cilíndricas, el elemento de área es $dA = r\,dr\,d\phi$. Al integrar sobre el radio, aparece el factor $r$. Por ende, las funciones de Bessel son ortogonales con peso **$w(r) = r$**.
   - En coordenadas esféricas, el elemento de ángulo sólido es $d\Omega = \sin\theta\,d\theta\,d\phi$. Haciendo el cambio $x = \cos\theta$, tenemos $dx = -\sin\theta\,d\theta$, lo que hace que los polinomios de Legendre tengan peso **$w(x) = 1$** en $x \in [-1, 1]$.

2. **Estados fundamentales cuánticos y distribuciones gaussianas:**
   - En el oscilador armónico cuántico, el estado fundamental es proporcional a una campana de Gauss $\psi_0(x) \propto e^{-x^2/2}$. Al calcular el producto interno de estados excitados $\psi_n = H_n(x) e^{-x^2/2}$, el producto $\psi_n(x)\psi_m(x)$ contiene el factor $e^{-x^2}$. Por eso los polinomios de Hermite requieren el peso **$w(x) = e^{-x^2}$**.
   - En el átomo de hidrógeno (potencial de Coulomb), las funciones de onda radiales decaen exponencialmente como $e^{-r/a_0}$, lo que da lugar al peso de Laguerre **$w(x) = x^\alpha e^{-x}$**.

```python
# ============================================================
# ⚖️ Comparativa Visual de las Funciones de Peso w(x)
# ============================================================

pesos = [
    ("Fourier / Legendre:  w(x) = 1",                  lambda x: np.ones_like(x),        (-1, 1),       '#58a6ff'),
    ("Chebyshev I:  w(x) = (1-x²)^(-1/2)",            lambda x: 1/np.sqrt(1 - x**2),    (-0.98, 0.98), '#3fb950'),
    ("Chebyshev II:  w(x) = √(1-x²)",                 lambda x: np.sqrt(1 - x**2),      (-1, 1),       '#d29922'),
    ("Hermite:  w(x) = exp(-x²)",                     lambda x: np.exp(-x**2),          (-3, 3),       '#f85149'),
    ("Laguerre:  w(x) = exp(-x)",                     lambda x: np.exp(-x),             (0, 5),        '#bc8cff'),
    ("Laguerre gral. (α=1):  w(x) = x exp(-x)",       lambda x: x * np.exp(-x),         (0, 6),        '#39c5cf'),
    ("Jacobi (α=β=1):  w(x) = (1-x)(1+x)",           lambda x: (1-x)*(1+x),            (-1, 1),       '#ff7b72'),
    ("Bessel:  w(x) = x",                            lambda x: x,                      (0, 4),        '#e3b341'),
]

fig, axes = plt.subplots(4, 2, figsize=(14, 11))
axes = axes.flatten()
fig.patch.set_facecolor('#0d1117')

for ax, (nombre, w_func, (a, b), col) in zip(axes, pesos):
    xx = np.linspace(a, b, 500)
    yy = w_func(xx)
    ax.plot(xx, yy, color=col, lw=2.5, label='w(x)')
    ax.fill_between(xx, 0, yy, color=col, alpha=0.18)
    ax.set_title(nombre, fontweight='bold', fontsize=11)
    ax.set_xlabel('x')
    ax.set_ylabel('w(x)')
    ax.grid(True)
    ax.legend(loc='upper right', framealpha=0.25)

plt.tight_layout()
plt.show()
```

## 7. 🔬 Ortogonalidad a la Par: Integración Simbólica Exacta (`SymPy`) vs. Cuadratura Numérica (`SciPy`)

El teorema central de Sturm–Liouville afirma que la **matriz de Gram** ponderada:

$$G_{nm} = \langle y_n, y_m \rangle_w = \int_a^b y_n(x)\,y_m(x)\,w(x)\,dx$$

es **estrictamente diagonal**:

$$G_{nm} = \begin{cases} \|y_n\|_w^2 > 0 & \text{si } n = m \\ 0 & \text{si } n \neq m \end{cases}$$

A continuación ejecutamos ambos enfoques a la par:
1. **Enfoque Simbólico Exacto:** `sympy.integrate` calcula la integral definida analítica, produciendo fracciones racionales exactas como $2/3, 2/5, 2/7$ y factores $\pi$.
2. **Enfoque Numérico:** `scipy.integrate.quad` calcula el valor en punto flotante con precisión de máquina ($10^{-16}$).

```python
# ============================================================
# 🔬 Cálculo a la par: Simbólico (SymPy) y Numérico (SciPy)
# ============================================================

N = 4  # Modos 0 a 3

print("=" * 85)
print("  MATRIZ DE GRAM PARA POLINOMIOS DE LEGENDRE P_n(x) EN [-1, 1]")
print("  Exacto: ∫_{-1}^1 P_n(x) P_m(x) dx = [2 / (2n + 1)] · δ_{nm}")
print("=" * 85)

G_simb = []
G_num = np.zeros((N, N))

for i in range(N):
    fila_simb = []
    for j in range(N):
        # 1. Cálculo simbólico exacto con SymPy
        val_sym = smp.integrate(smp.legendre(i, x) * smp.legendre(j, x), (x, -1, 1))
        fila_simb.append(val_sym)
        
        # 2. Cálculo numérico con SciPy quad
        integrando = lambda t: sp.eval_legendre(i, t) * sp.eval_legendre(j, t)
        val_num, _ = integrate.quad(integrando, -1, 1)
        G_num[i, j] = val_num
    G_simb.append(fila_simb)

print("\n--- TABLA COMPARATIVA SIMBÓLICA vs NUMÉRICA ---")
print(f"{'Par (n, m)':<12} | {'Simbólico (SymPy exacto)':<25} | {'Numérico (SciPy float)':<25} | {'Diferencia':<15}")
print("-" * 85)
for i in range(N):
    for j in range(N):
        val_s = G_simb[i][j]
        val_n = G_num[i, j]
        dif = abs(float(val_s) - val_n)
        print(f"({i}, {j}){' ':>6} | {str(val_s):<25} | {val_n:<25.8f} | {dif:<15.2e}")
print("=" * 85)

# --- Visualización gráfica de las Matrices de Gram (Heatmaps) ---
N_modos = 5
G_fourier = np.zeros((N_modos, N_modos))
for n_idx in range(1, N_modos + 1):
    for m_idx in range(1, N_modos + 1):
        val, _ = integrate.quad(lambda t: np.sin(n_idx*np.pi*t)*np.sin(m_idx*np.pi*t), 0, 1)
        G_fourier[n_idx-1, m_idx-1] = val

G_chebyshev = np.zeros((N_modos, N_modos))
for n_idx in range(N_modos):
    for m_idx in range(N_modos):
        # Transformación x = cos(θ) => ∫_0^π cos(n θ) cos(m θ) dθ
        val, _ = integrate.quad(lambda th: np.cos(n_idx*th)*np.cos(m_idx*th), 0, np.pi)
        G_chebyshev[n_idx, m_idx] = val

fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))
matrices = [
    (G_fourier, 'Fourier Seno [0, 1]', [f'n={k}' for k in range(1, 6)]),
    (G_num, 'Legendre P_n [-1, 1]', [f'P_{k}' for k in range(4)]),
    (G_chebyshev, 'Chebyshev T_n [-1, 1]', [f'T_{k}' for k in range(5)]),
]

for ax, (mat, titulo, etiquetas) in zip(axes, matrices):
    cax = ax.imshow(mat, cmap='viridis', interpolation='nearest')
    ax.set_title(f'Matriz de Gram: {titulo}', fontweight='bold')
    ax.set_xticks(range(len(etiquetas))); ax.set_xticklabels(etiquetas)
    ax.set_yticks(range(len(etiquetas))); ax.set_yticklabels(etiquetas)
    fig.colorbar(cax, ax=ax, fraction=0.046, pad=0.04)
    for r in range(mat.shape[0]):
        for c in range(mat.shape[1]):
            val = mat[r, c]
            color_txt = 'white' if val < mat.max()*0.6 else 'black'
            ax.text(c, r, f'{val:.3f}', ha='center', va='center',
                    color=color_txt, fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 8. 🎛️ Proyección Espectral: Coeficientes Analíticos Simbólicos vs. Reconstrucción Numérica

Debido a la **completez** del sistema de Sturm–Liouville, cualquier función cuadráticamente integrable $f(x) \in L^2_w([a, b])$ puede expandirse en una serie generalizada:

$$f(x) = \sum_{n=0}^{\infty} c_n\,y_n(x), \qquad c_n = \frac{\langle f, y_n \rangle_w}{\|y_n\|_w^2} = \frac{\int_a^b f(x)\,y_n(x)\,w(x)\,dx}{\int_a^b [y_n(x)]^2\,w(x)\,dx}$$

### Caso de estudio analítico con SymPy:
Consideremos la función parabólica $f(x) = x(1-x)$ en $[0, 1]$. Con `SymPy` podemos calcular la fórmula analítica cerrada de todos los coeficientes de Fourier seno:

$$c_n = 2\int_0^1 x(1-x)\sin(n\pi x)\,dx = \begin{cases} \frac{8}{\pi^3 n^3} & n \text{ impar} \\ 0 & n \text{ par} \end{cases}$$

Veámoslo a continuación en código simbólico y contrastemos con el widget interactivo.

```python
# ============================================================
# 📐 Cálculo Simbólico Exacto de los Coeficientes c_n
# ============================================================

f_simb = x * (1 - x)

print("=" * 75)
print(f"  FÓRMULA ANALÍTICA SIMBÓLICA CON SYMPY PARA f(x) = {f_simb}")
print("=" * 75)

for k in range(1, 7):
    # Integral analítica del numerador
    numerador = smp.integrate(f_simb * smp.sin(k * smp.pi * x), (x, 0, 1))
    # Norma cuadrada ||sin(k pi x)||^2 = 1/2
    c_k_exacto = smp.simplify(numerador / smp.Rational(1, 2))
    c_k_num = float(c_k_exacto)
    print(f"  Modo n={k}: c_{k} = {str(c_k_exacto):<20}  ≈ {c_k_num:.8f}")

print("=" * 75)
print("💡 Nota pedagógica: Los coeficientes para n par son exactamente 0")
print("   debido a la simetría de f(x) alrededor del punto medio x = 1/2.")
```

```python
# ============================================================
# 🎛️ Widget interactivo: Aproximación en base ortogonal
# ============================================================

import ipywidgets as widgets
from IPython.display import display

def aproximar_serie(base='Fourier Seno (en [0, 1])', funcion_tipo='Pulso Cuadrado', N_terminos=5):
    """
    Calcula la serie generalizada truncada a N_terminos y grafica la aproximación.
    """
    plt.figure(figsize=(12, 5))

    if base == 'Fourier Seno (en [0, 1])':
        a, b = 0.0, 1.0
        xx = np.linspace(a, b, 500)
        
        if funcion_tipo == 'Pulso Cuadrado':
            f_eval = lambda t: np.where((t > 0.25) & (t < 0.75), 1.0, 0.0)
        elif funcion_tipo == 'Onda Triangular':
            f_eval = lambda t: np.where(t < 0.5, 2.0*t, 2.0*(1.0 - t))
        else:  # Parábola x*(1-x)
            f_eval = lambda t: 4.0 * t * (1.0 - t)

        yy_target = f_eval(xx)
        yy_aprox = np.zeros_like(xx)
        coefs = []

        for k in range(1, N_terminos + 1):
            val, _ = integrate.quad(lambda t: f_eval(t) * np.sin(k * np.pi * t), a, b)
            c_k = 2.0 * val
            coefs.append(c_k)
            yy_aprox += c_k * np.sin(k * np.pi * xx)

    else:  # Legendre (en [-1, 1])
        a, b = -1.0, 1.0
        xx = np.linspace(a, b, 500)

        if funcion_tipo == 'Pulso Cuadrado':
            f_eval = lambda t: np.where(t > 0.0, 1.0, -1.0)
        elif funcion_tipo == 'Onda Triangular':
            f_eval = lambda t: 1.0 - np.abs(t)
        else:  # Gaussiana
            f_eval = lambda t: np.exp(-3.0 * t**2)

        yy_target = f_eval(xx)
        yy_aprox = np.zeros_like(xx)
        coefs = []

        for k in range(N_terminos):
            val, _ = integrate.quad(lambda t: f_eval(t) * sp.eval_legendre(k, t), a, b)
            norm_sq = 2.0 / (2.0 * k + 1.0)
            c_k = val / norm_sq
            coefs.append(c_k)
            yy_aprox += c_k * sp.eval_legendre(k, xx)

    # --- Gráfica comparativa ---
    plt.subplot(1, 2, 1)
    plt.plot(xx, yy_target, 'w--', lw=2.2, label='Función objetivo $f(x)$')
    plt.plot(xx, yy_aprox, color='#58a6ff', lw=2.5, label=f'Aproximación $S_{{{N_terminos}}}(x)$')
    plt.title(f'{base} — {funcion_tipo}', fontweight='bold')
    plt.xlabel('x'); plt.ylabel('y')
    plt.legend(fontsize=10, framealpha=0.3); plt.grid(True)

    # --- Espectro de coeficientes c_n ---
    plt.subplot(1, 2, 2)
    indices = np.arange(1, len(coefs)+1) if 'Fourier' in base else np.arange(len(coefs))
    plt.bar(indices, np.abs(coefs), color='#3fb950', alpha=0.85, edgecolor='white')
    plt.title(r'Espectro de amplitudes $|c_n|$', fontweight='bold')
    plt.xlabel('Orden $n$'); plt.ylabel('$|c_n|$')
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

# Widgets interactivos
dropdown_base = widgets.Dropdown(
    options=['Fourier Seno (en [0, 1])', 'Legendre P_n (en [-1, 1])'],
    value='Fourier Seno (en [0, 1])',
    description='Base:'
)
dropdown_func = widgets.Dropdown(
    options=['Pulso Cuadrado', 'Onda Triangular', 'Parábola / Gaussiana'],
    value='Pulso Cuadrado',
    description='Función $f$:'
)
slider_n = widgets.IntSlider(
    min=1, max=25, step=1, value=5,
    description='Términos $N$:'
)

ui = widgets.VBox([widgets.HBox([dropdown_base, dropdown_func]), slider_n])
out = widgets.interactive_output(aproximar_serie, {
    'base': dropdown_base,
    'funcion_tipo': dropdown_func,
    'N_terminos': slider_n
})
display(ui, out)
print("🎬 Modifica el slider de Términos N para observar el fenómeno de Gibbs o la aceleración de convergencia.")
```

1. **Sinergia Simbólica y Numérica:** Con `SymPy` verificamos analíticamente que las autofunciones anulan idénticamente el operador de Sturm–Liouville $\mathcal{L}[y] - \lambda w y = 0$ y calculamos integrales exactas racionales. 2. **Unificación:** El análisis clásico de Fourier (seno, coseno, exponencial) es simplemente el problema de Sturm–Liouville más elemental con $p=1, q=0, w=1$. Las condiciones de frontera seleccionan la base apropiada.
3. **Origen de las funciones especiales:** Cuando cambiamos de geometría (cilíndrica $\to$ Bessel, esférica $\to$ Legendre) o introducimos confinamiento cuántico (Gaussiano $\to$ Hermite, Coulombiano $\to$ Laguerre), la ecuación diferencial adopta la forma canónica de Sturm–Liouville con una función de peso $w(x) \neq 1$.
4. **Ortogonalidad ponderada:** Todas las funciones especiales son autofunciones de un operador autoadjunto, lo que garantiza matemáticamente que la matriz de Gram sea estrictamente diagonal bajo la métrica ponderada $\langle f, g \rangle_w = \int_a^b f(x)g(x)w(x)\,dx$.
