## 1. El Problema General de Sturm–Liouville

En un intervalo $x \in [a, b]$, la **forma canónica o autoadjunta** de la ecuación diferencial de Sturm–Liouville es:

$$\boxed{\mathcal{L}[y](x) \equiv -\frac{d}{dx}\left[p(x)\frac{dy}{dx}\right] + q(x)y(x) = \lambda w(x)y(x)}$$

o bien, expandiendo la derivada exterior:

$$\left[p(x)y'\right]' + \big(\lambda w(x) - q(x)\big)y = 0$$

### 1.1 Elementos del problema
- $p(x) > 0$ en $(a, b)$: función de rigidez o coeficiente diferencial principal (continuamente diferenciable, $C^1$).
- $w(x) > 0$ en $(a, b)$: **función de peso** (*weight function*), continua y estrictamente positiva casi en todas partes.
- $q(x)$: función potencial real y continua en $[a, b]$.
- $\lambda$: parámetro escalar de **autovalor** (*eigenvalue*).
- $y(x)$: **autofunción** (*eigenfunction*) asociada al autovalor $\lambda$.

### 1.2 Condiciones de frontera regulares (de tipo Sturm–Liouville)
Para garantizar que el operador diferencial $\mathcal{L}$ sea **formalmente autoadjunto** (simétrico respecto al producto interno en $L^2$), se imponen condiciones lineales homogéneas en los extremos:

$$\alpha_1 y(a) + \alpha_2 p(a) y'(a) = 0, \qquad \beta_1 y(b) + \beta_2 p(b) y'(b) = 0$$

donde $(\alpha_1, \alpha_2) \neq (0,0)$ y $(\beta_1, \beta_2) \neq (0,0)$.

### 1.3 Teoremas fundamentales de Sturm–Liouville
Para un problema de Sturm–Liouville regular:
1. **Realidad del espectro:** Todos los autovalores $\lambda_n$ son números reales.
2. **Espectro discreto y ordenado:** Existe una cantidad infinita contable de autovalores acotados inferiormente:
   $$\lambda_0 < \lambda_1 < \lambda_2 < \dots < \lambda_n < \dots \to +\infty$$
3. **Teorema de oscilación:** La autofunción $y_n(x)$ asociada a $\lambda_n$ tiene exactamente $n$ ceros simples en el intervalo abierto $(a, b)$.
4. **Ortogonalidad ponderada:** Si $\lambda_n \neq \lambda_m$, las autofunciones $y_n(x)$ e $y_m(x)$ son ortogonales con respecto al peso $w(x)$:
   $$\langle y_n, y_m \rangle_w \equiv \int_a^b y_n(x)\,y_m(x)\,w(x)\,dx = \delta_{nm}\,N_n$$
   donde $\delta_{nm}$ es la delta de Kronecker y $N_n = \|y_n\|_w^2$ es la norma cuadrática ponderada.
5. **Completitud:** Las autofunciones $\{y_n(x)\}_{n=0}^\infty$ forman una **base ortogonal completa** del espacio de Hilbert ponderado $L^2_w([a,b])$.

## 2. Fourier: El Caso Prototípico Fundamental

Una de las conexiones conceptuales más profundas y hermosas de la física matemática es entender que **el análisis de Fourier no es un tema aislado**, sino **el caso regular más elemental posible de Sturm–Liouville**:

$$p(x) = 1, \qquad q(x) = 0, \qquad w(x) = 1$$

Sustituyendo estos coeficientes en la ecuación general obtenemos la ecuación del oscilador armónico espacial:

$$\mathcal{L}[y] = -\frac{d}{dx}\left[1\cdot \frac{dy}{dx}\right] + 0 = -y'' = \lambda y$$

Es decir:

$$y''(x) + \lambda y(x) = 0$$

### ¿De dónde salen los senos, los cosenos o ambos?
La ecuación diferencial es idéntica en todos los casos; **lo que cambia radicalmente la base de autofunciones son las condiciones de frontera** impuestas por la física del problema:

| Condición de frontera | Dominio | Autofunciones $y_n(x)$ | Autovalores $\lambda_n$ | Interpretación en Física |
|:---|:---:|:---|:---:|:---|
| **Dirichlet** (Frontera fija) | $[0, L]$ | $\sin\left(\frac{n\pi x}{L}\right), \quad n=1,2,\dots$ | $\left(\frac{n\pi}{L}\right)^2$ | **Partícula en una caja cuántica 1D** ($V=\infty$ en bordes); cuerda de guitarra con extremos fijos. |
| **Neumann** (Derivada nula) | $[0, L]$ | $\cos\left(\frac{n\pi x}{L}\right), \quad n=0,1,\dots$ | $\left(\frac{n\pi}{L}\right)^2$ | Barra con extremos perfectamente aislados térmicamente (flujo de calor $\partial T/\partial x = 0$); extremos libres. |
| **Periódicas** | $[-L, L]$ | $\left\{1, \cos\frac{n\pi x}{L}, \sin\frac{n\pi x}{L}\right\}$ o $\{e^{in\pi x/L}\}$ | $\left(\frac{n\pi}{L}\right)^2$ | Anillo unidimensional cerrado; electrón en un potencial periódico de celda cristalina (teorema de Bloch). |

> 💡 **Conclusión clave:** La partícula en una caja que estudiaste en física cuántica o la cuerda clásica fija **es** exactamente el problema de Sturm–Liouville con peso unitario y condiciones Dirichlet. Cuando decimos "desarrollo en serie de Fourier", simplemente estamos aplicando el **teorema de completez de Sturm–Liouville** para el operador $-d^2/dx^2$.

## 3. 📋 La Gran Tabla de Sturm–Liouville y Funciones Especiales

Cuando el dominio no es plano o uniforme (por ejemplo en geometrías esféricas, cilíndricas o potenciales cuánticos gaussianos), los coeficientes métricos transforman $p(x)$, $q(x)$ y el peso $w(x)$. A continuación se presenta la tabla canónica que unifica todas las grandes familias ortogonales de la física:

| Nombre | Dominio $[a,b]$ | Peso $w(x)$ | Primeras funciones ortogonales | Forma Sturm–Liouville |
|:---|:---:|:---:|:---|:---|
| **Fourier (completa, real)** | $[-L, L]$ | $1$ | $1,\;\cos\frac{\pi x}{L},\;\sin\frac{\pi x}{L},\;\cos\frac{2\pi x}{L},\;\sin\frac{2\pi x}{L},\dots$ | $-y'' = \lambda y,\quad \lambda_n=\left(\frac{n\pi}{L}\right)^2$ |
| **Fourier (compleja)** | $[-L, L]$ | $1$ | $e^{in\pi x/L},\; n\in\mathbb{Z}$ | $-y'' = \lambda y,\quad \lambda_n=\left(\frac{n\pi}{L}\right)^2$ |
| **Fourier seno (half-range)** | $[0, L]$ | $1$ | $\sin\frac{\pi x}{L},\;\sin\frac{2\pi x}{L},\;\sin\frac{3\pi x}{L},\dots$ | $-y'' = \lambda y,\; y(0)=y(L)=0$ |
| **Fourier coseno (half-range)** | $[0, L]$ | $1$ | $1,\;\cos\frac{\pi x}{L},\;\cos\frac{2\pi x}{L},\dots$ | $-y'' = \lambda y,\; y'(0)=y'(L)=0$ |
| **Legendre** $P_n(x)$ | $[-1, 1]$ | $1$ | $P_0=1,\; P_1=x,\; P_2=\frac{1}{2}(3x^2-1),\; P_3=\frac{1}{2}(5x^3-3x)$ | $\left[(1-x^2)P_n'\right]' + n(n+1)P_n = 0$ |
| **Chebyshev I** $T_n(x)$ | $[-1, 1]$ | $(1-x^2)^{-1/2}$ | $T_0=1,\; T_1=x,\; T_2=2x^2-1,\; T_3=4x^3-3x$ | $\left[(1-x^2)^{1/2}T_n'\right]' + \frac{n^2}{(1-x^2)^{1/2}}T_n = 0$ |
| **Chebyshev II** $U_n(x)$ | $[-1, 1]$ | $(1-x^2)^{1/2}$ | $U_0=1,\; U_1=2x,\; U_2=4x^2-1,\; U_3=8x^3-4x$ | $\left[(1-x^2)^{1/2}U_n'\right]' + \frac{n(n+2)}{(1-x^2)^{1/2}}U_n = 0$ |
| **Hermite** $H_n(x)$ | $(-\infty, \infty)$ | $e^{-x^2}$ | $H_0=1,\; H_1=2x,\; H_2=4x^2-2,\; H_3=8x^3-12x$ | $\left[e^{-x^2}H_n'\right]' + 2n e^{-x^2}H_n = 0$ |
| **Laguerre** $L_n(x)$ | $[0, \infty)$ | $e^{-x}$ | $L_0=1,\; L_1=1-x,\; L_2=1-2x+\frac{1}{2}x^2,\; L_3=1-3x+\frac{3}{2}x^2-\frac{1}{6}x^3$ | $\left[x e^{-x}L_n'\right]' + n e^{-x}L_n = 0$ |
| **Laguerre generalizado** $L_n^{(\alpha)}(x)$ | $[0, \infty)$ | $x^\alpha e^{-x},\; \alpha>-1$ | $L_0^{(\alpha)}=1,\; L_1^{(\alpha)}=1+\alpha-x,\; L_2^{(\alpha)}=\frac{1}{2}(x^2-2(\alpha+2)x+(\alpha+1)(\alpha+2))$ | $\left[x^{\alpha+1}e^{-x}(L_n^{(\alpha)})'\right]' + n x^\alpha e^{-x}L_n^{(\alpha)} = 0$ |
| **Jacobi** $P_n^{(\alpha,\beta)}(x)$ | $[-1, 1]$ | $(1-x)^\alpha(1+x)^\beta$ | $P_0^{(\alpha,\beta)}=1,\; P_1^{(\alpha,\beta)}=\frac{1}{2}(\alpha+\beta+2)x+\frac{1}{2}(\alpha-\beta)$ | $\left[(1-x^2)w(P_n^{(\alpha,\beta)})'\right]' + n(n+\alpha+\beta+1)wP_n^{(\alpha,\beta)} = 0$ |
| **Gegenbauer** $C_n^{(\alpha)}(x)$ | $[-1, 1]$ | $(1-x^2)^{\alpha-1/2},\; \alpha>-1/2$ | $C_0^{(\alpha)}=1,\; C_1^{(\alpha)}=2\alpha x,\; C_2^{(\alpha)}=2\alpha(1+\alpha)x^2-\alpha$ | $\left[(1-x^2)^{\alpha+1/2}(C_n^{(\alpha)})'\right]' + n(n+2\alpha)w C_n^{(\alpha)} = 0$ |
| **Bessel** $J_\nu(k_n x)$ | $[0, a]$ | $x$ | $J_0(k_1 x),\; J_1(k_1 x),\; J_0(k_2 x),\; J_1(k_2 x), \dots$ | $\left[x J_\nu'\right]' + \left(k^2 x - \frac{\nu^2}{x}\right)J_\nu = 0$ |

> 🔍 **Observación sobre problemas singulares:** Cuando el intervalo es infinito (como en Hermite o Laguerre) o cuando $p(x)=0$ en algún extremo (como en Legendre donde $p(\pm 1)=1-(\pm 1)^2=0$), el problema se denomina **singular**. En estos casos, las condiciones de frontera regulares se sustituyen por la exigencia física de que la solución sea **regular y de cuadrado integrable** en los extremos.

## 5. ⚖️ El Papel Físico y Matemático de la Función de Peso $w(x)$

¿Por qué no todas las familias tienen peso uniforme $w(x)=1$?

En álgebra lineal y mecánica cuántica, la ortogonalidad estándar $\int f(x)g(x)\,dx = 0$ asume una medida de Lebesgue plana. Sin embargo, en física casi nunca trabajamos en espacios unidimensionales planos sin potencial:

1. **Cambio de coordenadas (Factores de escala o Jacobiano):**
   - En coordenadas cilíndricas, el elemento de área es $dA = r\,dr\,d\phi$. Al integrar sobre el radio, aparece el factor $r$. Por ende, las funciones de Bessel son ortogonales con peso **$w(r) = r$**.
   - En coordenadas esféricas, el elemento de ángulo sólido es $d\Omega = \sin\theta\,d\theta\,d\phi$. Haciendo el cambio $x = \cos\theta$, tenemos $dx = -\sin\theta\,d\theta$, lo que hace que los polinomios de Legendre tengan peso **$w(x) = 1$** en $x \in [-1, 1]$.

2. **Estados fundamentales cuánticos y distribuciones gaussianas:**
   - En el oscilador armónico cuántico, el estado fundamental es proporcional a una campana de Gauss $\psi_0(x) \propto e^{-x^2/2}$. Al calcular el producto interno de estados excitados $\psi_n = H_n(x) e^{-x^2/2}$, el producto $\psi_n(x)\psi_m(x)$ contiene el factor $e^{-x^2}$. Por eso los polinomios de Hermite requieren el peso **$w(x) = e^{-x^2}$**.
   - En el átomo de hidrógeno (potencial de Coulomb), las funciones de onda radiales decaen exponencialmente como $e^{-r/a_0}$, lo que da lugar al peso de Laguerre **$w(x) = x^\alpha e^{-x}$**.

```python
# ============================================================
# ⚖️ Comparativa Visual de las Funciones de Peso w(x)
# ============================================================

pesos = [
    ("Fourier / Legendre:  w(x) = 1",                  lambda x: np.ones_like(x),        (-1, 1),       '#58a6ff'),
    ("Chebyshev I:  w(x) = (1-x²)^(-1/2)",            lambda x: 1/np.sqrt(1 - x**2),    (-0.98, 0.98), '#3fb950'),
    ("Chebyshev II:  w(x) = √(1-x²)",                 lambda x: np.sqrt(1 - x**2),      (-1, 1),       '#d29922'),
    ("Hermite:  w(x) = exp(-x²)",                     lambda x: np.exp(-x**2),          (-3, 3),       '#f85149'),
    ("Laguerre:  w(x) = exp(-x)",                     lambda x: np.exp(-x),             (0, 5),        '#bc8cff'),
    ("Laguerre gral. (α=1):  w(x) = x exp(-x)",       lambda x: x * np.exp(-x),         (0, 6),        '#39c5cf'),
    ("Jacobi (α=β=1):  w(x) = (1-x)(1+x)",           lambda x: (1-x)*(1+x),            (-1, 1),       '#ff7b72'),
    ("Bessel:  w(x) = x",                            lambda x: x,                      (0, 4),        '#e3b341'),
]

fig, axes = plt.subplots(4, 2, figsize=(14, 11))
axes = axes.flatten()
fig.patch.set_facecolor('#0d1117')

for ax, (nombre, w_func, (a, b), col) in zip(axes, pesos):
    xx = np.linspace(a, b, 500)
    yy = w_func(xx)
    ax.plot(xx, yy, color=col, lw=2.5, label='w(x)')
    ax.fill_between(xx, 0, yy, color=col, alpha=0.18)
    ax.set_title(nombre, fontweight='bold', fontsize=11)
    ax.set_xlabel('x')
    ax.set_ylabel('w(x)')
    ax.grid(True)
    ax.legend(loc='upper right', framealpha=0.25)

plt.tight_layout()
plt.show()
```

## 6. 🔬 Verificación Numérica del Teorema de Ortogonalidad

El teorema central de Sturm–Liouville afirma que la **matriz de Gram** ponderada:

$$G_{nm} = \langle y_n, y_m \rangle_w = \int_a^b y_n(x)\,y_m(x)\,w(x)\,dx$$

debe ser **estrictamente diagonal**:

$$G_{nm} = \begin{cases} \|y_n\|_w^2 > 0 & \text{si } n = m \\ 0 & \text{si } n \neq m \end{cases}$$

Comprobemos esto numéricamente calculando las integrales mediante cuadratura numérica de alta precisión (`scipy.integrate.quad`) para tres familias fundamentales: **Fourier Seno**, **Legendre** y **Chebyshev de 1ª especie**.

```python
# ============================================================
# 🧮 Cálculo numérico de la Matriz de Gram G_{nm}
# ============================================================

N_modos = 5  # Evaluaremos modos del orden 0 a 4 (o 1 a 5)

# 1. Matriz de Gram para Fourier Seno en [0, 1] con w(x)=1
G_fourier = np.zeros((N_modos, N_modos))
for n in range(1, N_modos + 1):
    for m in range(1, N_modos + 1):
        integrando = lambda x: np.sin(n * np.pi * x) * np.sin(m * np.pi * x)
        val, _ = integrate.quad(integrando, 0, 1)
        G_fourier[n-1, m-1] = val

# 2. Matriz de Gram para Legendre P_n en [-1, 1] con w(x)=1
G_legendre = np.zeros((N_modos, N_modos))
for n in range(N_modos):
    for m in range(N_modos):
        integrando = lambda x: sp.eval_legendre(n, x) * sp.eval_legendre(m, x)
        val, _ = integrate.quad(integrando, -1, 1)
        G_legendre[n, m] = val

# 3. Matriz de Gram para Chebyshev T_n en [-1, 1] con w(x)=1/sqrt(1-x^2)
# Usamos la transformación trigonométrica x = cos(θ) => dx = -sin(θ)dθ
# para evitar la divergencia numérica integrable en los bordes ±1:
# ∫_{-1}^1 T_n(x) T_m(x) / √(1-x²) dx = ∫_0^π cos(n θ) cos(m θ) dθ
G_chebyshev = np.zeros((N_modos, N_modos))
for n in range(N_modos):
    for m in range(N_modos):
        integrando = lambda th: np.cos(n * th) * np.cos(m * th)
        val, _ = integrate.quad(integrando, 0, np.pi)
        G_chebyshev[n, m] = val

# --- Visualización como mapas de calor (Heatmaps) ---
fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))
matrices = [
    (G_fourier, 'Fourier Seno [0, 1]', [f'n={i}' for i in range(1, 6)]),
    (G_legendre, 'Legendre P_n [-1, 1]', [f'P_{i}' for i in range(5)]),
    (G_chebyshev, 'Chebyshev T_n [-1, 1]', [f'T_{i}' for i in range(5)]),
]

for ax, (mat, titulo, etiquetas) in zip(axes, matrices):
    cax = ax.imshow(mat, cmap='viridis', interpolation='nearest')
    ax.set_title(f'Matriz de Gram: {titulo}', fontweight='bold')
    ax.set_xticks(range(N_modos)); ax.set_xticklabels(etiquetas)
    ax.set_yticks(range(N_modos)); ax.set_yticklabels(etiquetas)
    fig.colorbar(cax, ax=ax, fraction=0.046, pad=0.04)
    # Añadir valores numéricos dentro de cada celda
    for i in range(N_modos):
        for j in range(N_modos):
            val = mat[i, j]
            color_txt = 'white' if val < mat.max()*0.6 else 'black'
            ax.text(j, i, f'{val:.3f}', ha='center', va='center',
                    color=color_txt, fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()

print("💡 Observa que todos los elementos fuera de la diagonal son EXACTAMENTE 0.000.")
print("   Esto confirma numéricamente que las funciones son mutuamente ortogonales respecto a w(x).")
```

## 7. 🎛️ Explorador Interactivo: Expansión en Series Generalizadas de Fourier

Debido a la **completez** del sistema de Sturm–Liouville, cualquier función cuadráticamente integrable $f(x) \in L^2_w([a, b])$ puede expandirse en una **serie generalizada de Fourier**:

$$f(x) = \sum_{n=0}^{\infty} c_n\,y_n(x)$$

Multiplicando por $y_m(x)\,w(x)$ e integrando a ambos lados, la ortogonalidad aniquila todos los términos excepto cuando $n = m$:

$$\int_a^b f(x)\,y_m(x)\,w(x)\,dx = \sum_{n=0}^{\infty} c_n \underbrace{\int_a^b y_n(x)\,y_m(x)\,w(x)\,dx}_{\delta_{nm} \|y_m\|_w^2} = c_m\, \|y_m\|_w^2$$

Despejando el coeficiente de proyección generalizado:

$$\boxed{c_n = \frac{\langle f, y_n \rangle_w}{\|y_n\|_w^2} = \frac{\int_a^b f(x)\,y_n(x)\,w(x)\,dx}{\int_a^b [y_n(x)]^2\,w(x)\,dx}}$$

En la siguiente celda puedes seleccionar de forma interactiva la base ortogonal (Fourier seno o polinomios de Legendre), la función objetivo a aproximar y el número de términos $N$ para ver en tiempo real cómo la suma parcial $S_N(x)$ converge a $f(x)$.

```python
# ============================================================
# 🎛️ Widget interactivo: Aproximación en base ortogonal
# ============================================================

import ipywidgets as widgets
from IPython.display import display

def aproximar_serie(base='Fourier Seno', funcion_tipo='Pulso Cuadrado', N_terminos=5):
    """
    Calcula la serie generalizada truncada a N_terminos y grafica la aproximación.
    """
    plt.figure(figsize=(12, 5))

    if base == 'Fourier Seno (en [0, 1])':
        a, b = 0.0, 1.0
        xx = np.linspace(a, b, 500)
        
        # Función objetivo f(x)
        if funcion_tipo == 'Pulso Cuadrado':
            f_eval = lambda x: np.where((x > 0.25) & (x < 0.75), 1.0, 0.0)
        elif funcion_tipo == 'Onda Triangular':
            f_eval = lambda x: np.where(x < 0.5, 2.0*x, 2.0*(1.0 - x))
        else:  # Parábola
            f_eval = lambda x: 4.0 * x * (1.0 - x)

        yy_target = f_eval(xx)
        yy_aprox = np.zeros_like(xx)
        coefs = []

        for n in range(1, N_terminos + 1):
            # Coeficiente c_n = 2 * ∫_0^1 f(x) sin(n pi x) dx
            integrando = lambda x: f_eval(x) * np.sin(n * np.pi * x)
            val, _ = integrate.quad(integrando, a, b)
            c_n = 2.0 * val  # porque ||sin(n pi x)||^2 = 1/2
            coefs.append(c_n)
            yy_aprox += c_n * np.sin(n * np.pi * xx)

    else:  # Legendre (en [-1, 1])
        a, b = -1.0, 1.0
        xx = np.linspace(a, b, 500)

        if funcion_tipo == 'Pulso Cuadrado':
            f_eval = lambda x: np.where(x > 0.0, 1.0, -1.0)
        elif funcion_tipo == 'Onda Triangular':
            f_eval = lambda x: 1.0 - np.abs(x)
        else:  # Parábola / Gaussiana
            f_eval = lambda x: np.exp(-3.0 * x**2)

        yy_target = f_eval(xx)
        yy_aprox = np.zeros_like(xx)
        coefs = []

        for n in range(N_terminos):
            # Coeficiente c_n = [(2n+1)/2] * ∫_{-1}^1 f(x) P_n(x) dx
            integrando = lambda x: f_eval(x) * sp.eval_legendre(n, x)
            val, _ = integrate.quad(integrando, a, b)
            norm_sq = 2.0 / (2.0 * n + 1.0)
            c_n = val / norm_sq
            coefs.append(c_n)
            yy_aprox += c_n * sp.eval_legendre(n, xx)

    # --- Gráfica comparativa ---
    plt.subplot(1, 2, 1)
    plt.plot(xx, yy_target, 'w--', lw=2.2, label='Función objetivo $f(x)$')
    plt.plot(xx, yy_aprox, color='#58a6ff', lw=2.5, label=f'Aproximación $S_{{{N_terminos}}}(x)$')
    plt.title(f'{base} — {funcion_tipo}', fontweight='bold')
    plt.xlabel('x'); plt.ylabel('y')
    plt.legend(fontsize=10, framealpha=0.3); plt.grid(True)

    # --- Espectro de coeficientes c_n ---
    plt.subplot(1, 2, 2)
    indices = np.arange(1, len(coefs)+1) if 'Fourier' in base else np.arange(len(coefs))
    plt.bar(indices, np.abs(coefs), color='#3fb950', alpha=0.85, edgecolor='white')
    plt.title(r'Espectro de amplitudes $|c_n|$', fontweight='bold')
    plt.xlabel('Orden $n$'); plt.ylabel('$|c_n|$')
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

# Widgets interactivos
dropdown_base = widgets.Dropdown(
    options=['Fourier Seno (en [0, 1])', 'Legendre P_n (en [-1, 1])'],
    value='Fourier Seno (en [0, 1])',
    description='Base:'
)
dropdown_func = widgets.Dropdown(
    options=['Pulso Cuadrado', 'Onda Triangular', 'Parábola / Gaussiana'],
    value='Pulso Cuadrado',
    description='Función $f$:'
)
slider_n = widgets.IntSlider(
    min=1, max=25, step=1, value=5,
    description='Términos $N$:'
)

ui = widgets.VBox([widgets.HBox([dropdown_base, dropdown_func]), slider_n])
out = widgets.interactive_output(aproximar_serie, {
    'base': dropdown_base,
    'funcion_tipo': dropdown_func,
    'N_terminos': slider_n
})
display(ui, out)
print("🎬 Modifica el slider de Términos N para observar el fenómeno de Gibbs o la aceleración de convergencia.")
```

## 8. Conclusiones y Ejercicios Sugeridos

### Lo que aprendimos
1. **Unificación:** El análisis clásico de Fourier (seno, coseno, exponencial) es simplemente el problema de Sturm–Liouville más simple posible con $p=1, q=0, w=1$. Las condiciones de frontera seleccionan la base apropiada.
2. **El origen de las funciones especiales:** Cuando cambiamos de geometría (cilíndrica $\to$ Bessel, esférica $\to$ Legendre) o introducimos confinamiento cuántico (Gaussiano $\to$ Hermite, Coulombiano $\to$ Laguerre), la ecuación de onda o de Schrödinger se transforma en una versión singular de Sturm–Liouville con una función de peso $w(x) \neq 1$.
3. **Ortogonalidad ponderada:** Todas las funciones especiales son autofunciones de un operador autoadjunto, lo que garantiza matemáticamente que sean ortogonales bajo la métrica $\langle f, g \rangle_w = \int_a^b f(x)g(x)w(x)\,dx$.
4. **Completez y aproximación espectral:** Cualquier señal o estado cuántico cuadráticamente integrable puede descomponerse en estas bases. La proyección óptima se obtiene mediante los coeficientes generalizados de Fourier $c_n = \langle f, y_n \rangle_w / \|y_n\|_w^2$.
