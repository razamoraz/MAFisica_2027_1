## 🌀 PARTE I — El Oscilador Armónico Cuántico Unidimensional

Para aplicar la teoría de Sturm-Liouville al oscilador armónico cuántico unidimensional, transformaremos la ecuación de Schrödinger independiente del tiempo a la forma canónica autoadjunta. Dado que el dominio espacial es $x \\in (-\\infty, \\infty)$, este constituye un **problema de Sturm-Liouville singular**.

Recordemos la forma canónica de Sturm-Liouville:

$$\\frac{d}{dx}\\left[p(x)\\frac{dy}{dx}\\right] + q(x)y + \\lambda w(x)y = 0$$

A continuación, desarrollamos el procedimiento analítico estándar paso a paso.

---

### 1. Adimensionalización de la Ecuación de Schrödinger

La ecuación de Schrödinger para el oscilador armónico 1D es:

$$-\\frac{\\hbar^2}{2m} \\frac{d^2\\psi}{dx^2} + \\frac{1}{2}m\\omega^2 x^2 \\psi = E\\psi$$

Para eliminar las constantes físicas dimensionales $(\\hbar, m, \\omega)$, introducimos la variable espacial adimensional $\\xi$ y el parámetro de energía adimensional $\\epsilon$:

$$\\xi = \\sqrt{\\frac{m\\omega}{\\hbar}}\\,x, \\qquad \\epsilon = \\frac{2E}{\\hbar\\omega}$$

Calculando las derivadas respecto a $x$ mediante la regla de la cadena:

$$\\frac{d\\psi}{dx} = \\frac{d\\psi}{d\\xi}\\frac{d\\xi}{dx} = \\sqrt{\\frac{m\\omega}{\\hbar}}\\,\\frac{d\\psi}{d\\xi}, \\qquad \\frac{d^2\\psi}{dx^2} = \\frac{m\\omega}{\\hbar}\\,\\frac{d^2\\psi}{d\\xi^2}$$

Sustituyendo estas expresiones en la ecuación de Schrödinger y multiplicando por $-\\frac{2}{\\hbar\\omega}$:

$$\\frac{d^2\\psi}{d\\xi^2} + (\\epsilon - \\xi^2)\\psi = 0$$

En este punto, la ecuación ya posee la forma de Sturm-Liouville con $p(\\xi) = 1$, $q(\\xi) = -\\xi^2$, función de peso $w(\\xi) = 1$ y autovalor $\\epsilon$. Sin embargo, resolverla directamente en esta forma no revela de manera inmediata la base polinomial ortogonal ni la condición de acotamiento físico.

---

### 2. Extracción del Comportamiento Asintótico

Para que el estado represente una partícula cuántica ligada físicamente admisible, la función de onda debe ser de cuadrado integrable en todo el eje real:

$$\\lim_{|\\xi| \\to \\infty} \\psi(\\xi) = 0, \\qquad \\int_{-\\infty}^\\infty |\\psi(\\xi)|^2\\,d\\xi = 1$$

Analizando el régimen asintótico distante donde $\\xi^2 \\gg \\epsilon$, la ecuación diferencial se aproxima a:

$$\\frac{d^2\\psi}{d\\xi^2} \\approx \\xi^2 \\psi$$

Probando una solución exponencial de la forma $\\psi \\sim e^{\\pm \\xi^2 / 2}$:

$$\\frac{d}{d\\xi}\\left(e^{\\pm \\xi^2/2}\\right) = \\pm \\xi e^{\\pm \\xi^2/2}, \\qquad \\frac{d^2}{d\\xi^2}\\left(e^{\\pm \\xi^2/2}\\right) = (\\xi^2 \\pm 1)e^{\\pm \\xi^2/2} \\approx \\xi^2 e^{\\pm \\xi^2/2} \\quad (\\text{para } \\xi \\to \\infty)$$

La solución con signo positivo diverge catastróficamente ($e^{+\\xi^2/2} \\to \\infty$). Por lo tanto, la física exige la rama cuadráticamente integrable $e^{-\\xi^2/2}$.

Proponemos entonces el **ansatz factorizado**:

$$\\psi(\\xi) = H(\\xi)\\,e^{-\\xi^2/2}$$

Derivamos aplicando la regla del producto:

$$\\frac{d\\psi}{d\\xi} = \\left( \\frac{dH}{d\\xi} - \\xi H \\right) e^{-\\xi^2/2}$$

$$\\frac{d^2\\psi}{d\\xi^2} = \\left( \\frac{d^2H}{d\\xi^2} - 2\\xi\\frac{dH}{d\\xi} + (\\xi^2 - 1)H \\right) e^{-\\xi^2/2}$$

Sustituyendo $\\psi''$ en la ecuación diferencial adimensional $\\psi'' + (\\epsilon - \\xi^2)\\psi = 0$ y cancelando el factor gaussiano $e^{-\\xi^2/2} \\neq 0$:

$$\\frac{d^2H}{d\\xi^2} - 2\\xi\\frac{dH}{d\\xi} + (\\xi^2 - 1)H + (\\epsilon - \\xi^2)H = 0$$

Los términos cúbicos en $\\xi^2$ se cancelan con exactitud, dando lugar a la célebre **Ecuación Diferencial de Hermite**:

$$\\boxed{\\frac{d^2H}{d\\xi^2} - 2\\xi\\frac{dH}{d\\xi} + (\\epsilon - 1)H = 0}$$

---

### 3. Transformación a la Forma Canónica de Sturm-Liouville

La ecuación de Hermite contiene un término de primera derivada $-2\\xi H'$, por lo que **no se encuentra en forma autoadjunta**. Para llevarla a la forma canónica de Sturm-Liouville, calculamos su factor integrante $\\mu(\\xi)$:

$$\\mu(\\xi) = \\exp\\left( \\int P(\\xi)\\,d\\xi \\right) = \\exp\\left( \\int -2\\xi\\,d\\xi \\right) = e^{-\\xi^2}$$

Multiplicando toda la ecuación de Hermite por $\\mu(\\xi) = e^{-\\xi^2}$:

$$e^{-\\xi^2} \\frac{d^2H}{d\\xi^2} - 2\\xi e^{-\\xi^2} \\frac{dH}{d\\xi} + (\\epsilon - 1)e^{-\\xi^2} H = 0$$

Reconociendo que los dos primeros términos corresponden a la derivada de un producto:

$$\\frac{d}{d\\xi}\\left( e^{-\\xi^2} \\frac{dH}{d\\xi} \\right) + (\\epsilon - 1) e^{-\\xi^2} H = 0$$

Comparando directamente con la definición canónica $\\frac{d}{d\\xi}\\left[p(\\xi)\\frac{dH}{d\\xi}\\right] + q(\\xi)H + \\lambda w(\\xi)H = 0$, identificamos sin ambigüedad los elementos del problema:

- **Función de rigidez:** $p(\\xi) = e^{-\\xi^2}$
- **Potencial modificado:** $q(\\xi) = 0$
- **Función de peso:** $w(\\xi) = e^{-\\xi^2}$
- **Autovalor de Sturm–Liouville:** $\\lambda = \\epsilon - 1$

---

### 4. Consecuencias Físicas de la Teoría de Sturm–Liouville

**1. Cuantización de la Energía (Espectro Discreto):**  
Para que las soluciones $H(\\xi)$ no diverjan asintóticamente como $e^{\\xi^2}$ en el infinito (lo cual arruinaría la normalización de la función de onda total $\\psi$), la solución en serie de potencias debe truncarse en un polinomio de grado finito $n$. La teoría exige que el autovalor sea un entero par no negativo:

$$\\lambda = 2n, \\qquad n = 0, 1, 2, \\dots$$

Puesto que $\\lambda = \\epsilon - 1 = \\frac{2E}{\\hbar\\omega} - 1$, obtenemos inmediatamente el espectro de energías cuantizadas de Planck-Einstein:

$$\\boxed{E_n = \\hbar\\omega\\left(n + \\frac{1}{2}\\right)}$$

y las soluciones polinomiales $H_n(\\xi)$ son los **Polinomios de Hermite**. La energía del punto cero $E_0 = \\frac{1}{2}\\hbar\\omega$ emerge de forma natural como consecuencia de las condiciones asintóticas en el infinito.

**2. Ortogonalidad Ponderada:**  
El teorema fundamental de Sturm-Liouville establece que autofunciones asociadas a distintos autovalores son estrictamente ortogonales respecto a la función de peso $w(\\xi) = e^{-\\xi^2}$:

$$\\int_{-\\infty}^{\\infty} H_n(\\xi)\\,H_m(\\xi)\\,e^{-\\xi^2}\\,d\\xi = \\sqrt{\\pi}\\,2^n\\,n!\\;\\delta_{nm}$$

*Conexión física:* Las funciones de onda normalizadas totales $\\psi_n(\\xi) = N_n H_n(\\xi)e^{-\\xi^2/2}$ (donde $N_n = (\\sqrt{\\pi}2^n n!)^{-1/2}$) satisfacen la ortonormalidad cuántica estándar sin peso:

$$\\int_{-\\infty}^\\infty \\psi_n(\\xi)\\,\\psi_m(\\xi)\\,d\\xi = N_n N_m \\int_{-\\infty}^\\infty H_n(\\xi)\\,H_m(\\xi)\\,e^{-\\xi^2}\\,d\\xi = \\delta_{nm}$$

El peso $w(\\xi) = e^{-\\xi^2}$ queda perfectamente distribuido al multiplicarse las dos envolventes gaussianas $e^{-\\xi^2/2} \\cdot e^{-\\xi^2/2} = e^{-\\xi^2}$.

**3. Completitud:**  
Las autofunciones del problema singular de Sturm-Liouville forman una base ortonormal completa del espacio de Hilbert $L^2(\\mathbb{R})$. Cualquier estado cuántico arbitrario $|\\Psi\\rangle$ puede expandirse en una superposición única de autoestados de energía: $|\\Psi\\rangle = \\sum_{n=0}^\\infty c_n |\\psi_n\\rangle$ con $\\sum |c_n|^2 = 1$.

```python
# ============================================================
# 🛠️ Configuración del entorno, importaciones y diseño visual
# ============================================================

import math
import numpy as np
import matplotlib.pyplot as plt
import scipy.special as sp
import scipy.integrate as integrate
import sympy as smp

# --- Estilo visual temático oscuro premium (MAF - UNAM) ---
plt.style.use('dark_background')
plt.rcParams.update({
    'font.size': 11,
    'axes.titlesize': 13,
    'axes.labelsize': 11,
    'figure.facecolor': '#0d1117',
    'axes.facecolor': '#161b22',
    'grid.color': '#30363d',
    'grid.alpha': 0.45,
    'savefig.facecolor': '#0d1117',
})

# Paleta de colores armónicos de alto contraste
colores = ['#58a6ff', '#f97583', '#d2a8ff', '#3fb950', '#d29922']

print("✅ Entorno de computación científica cargado exitosamente.")
print(f"   NumPy: {np.__version__} | SciPy: {sp.__name__} | SymPy: {smp.__version__}")
```

```python
# ============================================================
# 🔬 Verificación Simbólica Formal: Operador de Sturm-Liouville
#    para los Polinomios de Hermite con SymPy
# ============================================================
# Demostramos que: L[H_n](ξ) - λ_n · w(ξ) · H_n(ξ) ≡ 0
# donde:
#    L[y] = - d/dξ [ exp(-ξ²) · dy/dξ ]
#    w(ξ) = exp(-ξ²)
#    λ_n  = 2n
# ============================================================

xi_sym = smp.symbols('xi', real=True)

def operador_SL_hermite(y_expr, var=xi_sym):
    # Aplica formalmente el operador de Sturm-Liouville:
    # L[y] = - d/dξ [ p(ξ) * dy/dξ ] donde p(ξ) = exp(-ξ²)
    p_expr = smp.exp(-var**2)
    dy_dvar = smp.diff(y_expr, var)
    termino_flujo = p_expr * dy_dvar
    return -smp.diff(termino_flujo, var)

print("=" * 80)
print("  COMPROBACIÓN SIMBÓLICA CON SYMPY: L[H_n] - 2n · exp(-ξ²) · H_n == 0")
print("=" * 80)

for n_val in [0, 1, 2, 3, 4]:
    # Polinomio de Hermite analítico exacto
    H_n = smp.hermite(n_val, xi_sym)
    
    # Aplicación formal del operador diferencial
    L_Hn = operador_SL_hermite(H_n, var=xi_sym)
    
    # Lado derecho canónico: λ_n * w(ξ) * H_n(ξ)
    lambda_n = 2 * n_val
    w_expr = smp.exp(-xi_sym**2)
    lado_derecho = lambda_n * w_expr * H_n
    
    # Residuo analítico: debe simplificarse idénticamente a 0
    residuo = smp.simplify(L_Hn - lado_derecho)
    
    print(f"Modo n={n_val}:  H_{n_val}(ξ) = {str(H_n):<22} | λ_{n_val} = {lambda_n}")
    print(f"   L[H_{n_val}]        = {L_Hn}")
    print(f"   λ_{n_val}·w(ξ)·H_{n_val}  = {lado_derecho}")
    print(f"   Residuo analítico = {residuo}  ✓ (¡Cero exacto!)")
    print("-" * 80)

print("🎉 El operador de Sturm-Liouville anula idénticamente todos los modos de Hermite.")
```

```python
# ============================================================
# 📊 Visualización 2D: Polinomios de Hermite, Funciones de Onda
#    y Cascada de Niveles de Energía del Oscilador Armónico
# ============================================================

def psi_oac(n, xi):
    # Función de onda cuántica normalizada ψ_n(ξ) = N_n H_n(ξ) exp(-ξ²/2)
    norma = 1.0 / np.sqrt((2.0**n) * math.factorial(n) * np.sqrt(np.pi))
    return norma * sp.eval_hermite(n, xi) * np.exp(-xi**2 / 2.0)

xi_pts = np.linspace(-4.5, 4.5, 600)

fig, axes = plt.subplots(2, 2, figsize=(16, 11))
fig.patch.set_facecolor('#0d1117')

# ------------------------------------------------------------
# 1. Polinomios de Hermite H_n(ξ) sin modular
# ------------------------------------------------------------
ax1 = axes[0, 0]
for n in range(4):
    ax1.plot(xi_pts, sp.eval_hermite(n, xi_pts), label=rf'$H_{n}(\xi)$',
             color=colores[n], lw=2.4)
ax1.set_title(r'1. Polinomios de Hermite $H_n(\xi)$ (Sin acotar)', fontweight='bold')
ax1.set_xlabel(r'$\xi$'); ax1.set_ylabel(r'$H_n(\xi)$')
ax1.set_ylim(-15, 25); ax1.legend(loc='upper right', framealpha=0.3); ax1.grid(True)

# ------------------------------------------------------------
# 2. Funciones de onda normalizadas ψ_n(ξ)
# ------------------------------------------------------------
ax2 = axes[0, 1]
for n in range(4):
    ax2.plot(xi_pts, psi_oac(n, xi_pts), label=rf'$\psi_{n}(\xi)$ ($n={n}$)',
             color=colores[n], lw=2.4)
ax2.axhline(0, color='white', lw=0.6, alpha=0.3)
ax2.set_title(r'2. Autofunciones Cuánticas $\psi_n(\xi) = N_n H_n e^{-\xi^2/2}$', fontweight='bold')
ax2.set_xlabel(r'$\xi$'); ax2.set_ylabel(r'$\psi_n(\xi)$')
ax2.legend(loc='upper right', framealpha=0.3); ax2.grid(True)

# ------------------------------------------------------------
# 3. Cascada de Energía de Feynman: V(ξ) + |ψ_n(ξ)|²
# ------------------------------------------------------------
ax3 = axes[1, 0]
V_pot = 0.5 * xi_pts**2
ax3.plot(xi_pts, V_pot, color='white', lw=2.0, ls='--', label=r'Potencial $V(\xi) = \frac{1}{2}\xi^2$')

escala_densidad = 1.2
for n in range(5):
    E_n = n + 0.5  # Energía en unidades de ℏω
    densidad = np.abs(psi_oac(n, xi_pts))**2
    ax3.axhline(E_n, color=colores[n%len(colores)], lw=0.9, ls=':', alpha=0.6)
    ax3.plot(xi_pts, E_n + escala_densidad * densidad, color=colores[n%len(colores)], lw=2.2)
    ax3.fill_between(xi_pts, E_n, E_n + escala_densidad * densidad,
                     color=colores[n%len(colores)], alpha=0.25)
    ax3.text(3.6, E_n + 0.08, rf'$E_{n} = {n}+\frac{{1}}{{2}}$',
             color=colores[n%len(colores)], fontsize=10, fontweight='bold')

ax3.set_title(r'3. Cascada Cuántica: $E_n + |\psi_n(\xi)|^2$ en el Pozo Armónico', fontweight='bold')
ax3.set_xlabel(r'$\xi$'); ax3.set_ylabel(r'Energía $E / (\hbar\omega)$')
ax3.set_xlim(-4.5, 4.5); ax3.set_ylim(0, 5.8)
ax3.legend(loc='upper left', framealpha=0.3); ax3.grid(True)

# ------------------------------------------------------------
# 4. Función de peso w(ξ) = exp(-ξ²) y distribución de Gauss
# ------------------------------------------------------------
ax4 = axes[1, 1]
peso_gauss = np.exp(-xi_pts**2)
ax4.plot(xi_pts, peso_gauss, color='#f97583', lw=2.5, label=r'Peso $w(\xi) = e^{-\xi^2}$')
ax4.fill_between(xi_pts, 0, peso_gauss, color='#f97583', alpha=0.2)
ax4.plot(xi_pts, np.exp(-xi_pts**2 / 2.0), color='#58a6ff', lw=1.8, ls='-.',
         label=r'Envolvente $\sqrt{w(\xi)} = e^{-\xi^2/2}$')
ax4.set_title(r'4. Función de Peso Gaussiana $w(\xi)$ de Sturm–Liouville', fontweight='bold')
ax4.set_xlabel(r'$\xi$'); ax4.set_ylabel(r'$w(\xi)$')
ax4.legend(loc='upper right', framealpha=0.3); ax4.grid(True)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 🔬 Matriz de Gram: Ortogonalidad Ponderada de Hermite
#    Cálculo Numérico (SciPy) vs Analítico Exacto
# ============================================================
# Teorema S-L: ∫_{-∞}^∞ H_n(ξ) H_m(ξ) exp(-ξ²) dξ = √π · 2ⁿ · n! · δ_{nm}
# ============================================================

N_modos = 4
G_hermite = np.zeros((N_modos, N_modos))

print("=" * 85)
print("  TABLA COMPARATIVA: MATRIZ DE GRAM PONDERADA PARA POLINOMIOS DE HERMITE")
print("=" * 85)
print(f"{'Par (n, m)':<12} | {'Exacto: √π · 2ⁿ · n!':<25} | {'Numérico (SciPy quad)':<25} | {'Error Absoluto':<15}")
print("-" * 85)

for n in range(N_modos):
    for m in range(N_modos):
        val_num, _ = integrate.quad(
            lambda t: sp.eval_hermite(n, t) * sp.eval_hermite(m, t) * np.exp(-t**2),
            -np.inf, np.inf
        )
        G_hermite[n, m] = val_num
        
        val_exacto = np.sqrt(np.pi) * (2.0**n) * math.factorial(n) if n == m else 0.0
        error = abs(val_num - val_exacto)
        
        if n == m or n < m:
            print(f"({n}, {m}){' ':>6} | {val_exacto:<25.8f} | {val_num:<25.8f} | {error:<15.2e}")

print("=" * 85)

# Visualización Gráfica de la Matriz de Gram
fig, ax = plt.subplots(figsize=(6.5, 5))
fig.patch.set_facecolor('#0d1117')
cax = ax.imshow(G_hermite, cmap='plasma', interpolation='nearest')
ax.set_title(r'Matriz de Gram Ponderada $\langle H_n, H_m \rangle_{e^{-\xi^2}}$', fontweight='bold')
ax.set_xticks(range(N_modos)); ax.set_xticklabels([f'$H_{k}$' for k in range(N_modos)])
ax.set_yticks(range(N_modos)); ax.set_yticklabels([f'$H_{k}$' for k in range(N_modos)])
fig.colorbar(cax, ax=ax, fraction=0.046, pad=0.04)

for r in range(N_modos):
    for c in range(N_modos):
        val = G_hermite[r, c]
        color_txt = 'black' if val > G_hermite.max()*0.5 else 'white'
        ax.text(c, r, f'{val:.2f}', ha='center', va='center',
                color=color_txt, fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
print("💡 La matriz es estrictamente diagonal: ortogonalidad comprobada.")
```

## 🪐 PARTE II — El Átomo Hidrogenoide y los Polinomios de Laguerre

Para resolver el átomo de hidrógeno (o cualquier ion hidrogenoide con carga nuclear $Z$) mediante la teoría de Sturm-Liouville, nos enfocamos en la **ecuación radial**. En un potencial central de Coulomb $V(r) = -\\frac{Ze^2}{4\\pi\\epsilon_0 r}$, la función de onda espacial se factoriza en componentes angulares (Armónicos Esféricos $Y_l^m(\\theta, \\phi)$) y una parte radial $R(r)$:

$$\\psi_{nlm}(r, \\theta, \\phi) = R_{nl}(r)\\,Y_l^m(\\theta, \\phi)$$

El problema singular de Sturm-Liouville surge con total claridad en la determinación de esta función radial $R(r)$.

---

### 1. Adimensionalización de la Ecuación de Schrödinger Radial

La ecuación de Schrödinger radial para un electrón de masa reducida $\\mu$ en un potencial de Coulomb es:

$$-\\frac{\\hbar^2}{2\\mu} \\frac{1}{r^2} \\frac{d}{dr} \\left( r^2 \\frac{dR}{dr} \\right) + \\left[ -\\frac{Ze^2}{4\\pi\\epsilon_0 r} + \\frac{l(l+1)\\hbar^2}{2\\mu r^2} \\right] R = E R$$

Para los estados ligados ($E < 0$), introducimos un parámetro que absorbe la energía:

$$\\kappa = \\frac{\\sqrt{-2\\mu E}}{\\hbar}$$

y definimos la variable espacial adimensional $\\rho$:

$$\\rho = 2\\kappa r$$

Asimismo, agrupamos las constantes físicas en un parámetro adimensional $\\lambda$:

$$\\lambda = \\frac{Z e^2 \\mu}{4\\pi\\epsilon_0 \\hbar^2 \\kappa}$$

Sustituyendo $\\rho$ y $\\lambda$ en la ecuación diferencial y multiplicando por $-\\frac{2\\mu}{\\hbar^2 \\kappa^2}$, la ecuación radial colapsa a la forma adimensional:

$$\\frac{d^2 R}{d\\rho^2} + \\frac{2}{\\rho}\\frac{dR}{d\\rho} + \\left[ \\frac{\\lambda}{\\rho} - \\frac{1}{4} - \\frac{l(l+1)}{\\rho^2} \\right] R = 0$$

---

### 2. Extracción del Comportamiento Asintótico

Para encontrar una solución físicamente aceptable (normalizable en todo el espacio), analizamos los límites asintóticos extremos:

**Comportamiento en el infinito ($\\rho \\to \\infty$):**  
Los términos que dividen por $\\rho$ y $\\rho^2$ se vuelven despreciables. La ecuación se reduce asintóticamente a:

$$\\frac{d^2R}{d\\rho^2} - \\frac{1}{4}R \\approx 0 \\quad \\Longrightarrow \\quad R(\\rho) \\sim e^{-\\rho/2}$$

(descartando la rama físicamente divergente $e^{+\\rho/2}$).

**Comportamiento en el origen ($\\rho \\to 0$):**  
El término centrífugo domina la ecuación:

$$\\frac{d^2R}{d\\rho^2} + \\frac{2}{\\rho}\\frac{dR}{d\\rho} - \\frac{l(l+1)}{\\rho^2}R \\approx 0$$

Proponiendo $R \\sim \\rho^s$, obtenemos la ecuación indicial $s(s-1) + 2s - l(l+1) = s(s+1) - l(l+1) = 0$, cuyas raíces son $s = l$ y $s = -(l+1)$. La raíz negativa diverge en el origen, por lo que la solución físicamente admisible es:

$$R(\\rho) \\sim \\rho^l \\quad (\\text{para } \\rho \\to 0)$$

Para extraer ambos límites asintóticos, se propone el **ansatz factorizado**:

$$R(\\rho) = \\rho^l\\,e^{-\\rho/2}\\,L(\\rho)$$

Al calcular la primera y segunda derivada de $R(\\rho)$ y sustituirlas en la ecuación diferencial radial adimensional, obtenemos la ecuación diferencial para la función incógnita $L(\\rho)$:

$$\\rho \\frac{d^2L}{d\\rho^2} + (2l + 2 - \\rho) \\frac{dL}{d\\rho} + (\\lambda - l - 1) L = 0$$

Esta es con exactitud la célebre **Ecuación Diferencial Asociada de Laguerre**.

---

### 3. Transformación a la Forma de Sturm-Liouville

Al igual que ocurrió con la ecuación de Hermite, la ecuación de Laguerre no está en forma autoadjunta de manera explícita. Para hallar el factor integrante, dividimos la ecuación completa entre $\\rho$:

$$\\frac{d^2L}{d\\rho^2} + \\frac{2l+2-\\rho}{\\rho} \\frac{dL}{d\\rho} + \\frac{\\lambda-l-1}{\\rho} L = 0$$

El factor integrante $\\mu(\\rho)$ se calcula como:

$$\\mu(\\rho) = \\exp\\left( \\int \\frac{2l+2-\\rho}{\\rho} \\, d\\rho \\right) = \\exp\\big( (2l+2)\\ln\\rho - \\rho \\big) = \\rho^{2l+2}\\,e^{-\\rho}$$

Multiplicando toda la ecuación original por $\\rho^{2l+1} e^{-\\rho}$, la ecuación colapsa a una derivada exacta, adoptando la **forma canónica de Sturm-Liouville**:

$$\\boxed{\\frac{d}{d\\rho}\\left( \\rho^{2l+2} e^{-\\rho} \\frac{dL}{d\\rho} \\right) + (\\lambda - l - 1) \\rho^{2l+1} e^{-\\rho} L = 0}$$

Comparando término a término con $\\frac{d}{d\\rho}\\left[p(\\rho)\\frac{dL}{d\\rho}\\right] + q(\\rho)L + \\Lambda w(\\rho)L = 0$, identificamos rigurosamente los elementos del problema para los Polinomios Asociados de Laguerre:

- **Función $p(\\rho)$:** $\\rho^{2l+2} e^{-\\rho}$
- **Función $q(\\rho)$:** $0$
- **Función de peso $w(\\rho)$:** $\\rho^{2l+1} e^{-\\rho}$
- **Autovalor $\\Lambda$:** $\\lambda - l - 1$

---

### 4. Consecuencias Físicas de la Teoría

**1. Cuantización de la Energía (Espectro Discreto de Bohr):**  
Para que la solución $L(\\rho)$ sea regular en el infinito y la función de onda $R(\\rho)$ siga siendo integrable en todo el espacio, la serie de potencias que define a $L(\\rho)$ debe truncarse a un polinomio. La teoría de Sturm-Liouville exige que el autovalor $\\Lambda$ sea un entero no negativo:

$$\\Lambda = \\lambda - l - 1 = n_r \\quad \\implies \\quad \\lambda = n_r + l + 1$$

donde $n_r = 0, 1, 2, \\dots$ es el **número cuántico radial** (que cuenta el número de nodos esféricos interiores de la función de onda).

Definiendo el **número cuántico principal** como $n = n_r + l + 1$ (donde $n \\ge l + 1$), obtenemos $\\lambda = n$.  
Sustituyendo $\\lambda$ por su definición original en términos de la energía ($E = -\\frac{\\hbar^2 \\kappa^2}{2\\mu}$ con $\\kappa = \\frac{Z e^2 \\mu}{4\\pi\\epsilon_0 \\hbar^2 n}$), recuperamos con exactitud analítica el espectro energético de Bohr-Rydberg:

$$\\boxed{E_n = -\\frac{\\mu Z^2 e^4}{2(4\\pi\\epsilon_0)^2 \\hbar^2} \\frac{1}{n^2} = -\\frac{13.6\\,Z^2}{n^2}\\;\\text{eV}}$$

y las funciones resultantes $L_{n-l-1}^{2l+1}(\\rho)$ son los **Polinomios Asociados de Laguerre**.

**2. Ortogonalidad Ponderada:**  
El teorema de Sturm-Liouville garantiza que para un momento angular $l$ fijo, los polinomios de Laguerre con distintos números radiales son estrictamente ortogonales con respecto a la función de peso $w(\\rho) = \\rho^{2l+1} e^{-\\rho}$:

$$\\int_{0}^{\\infty} L_{n_r}^{2l+1}(\\rho)\\,L_{n'_r}^{2l+1}(\\rho) \\; \\rho^{2l+1} e^{-\\rho} \\, d\\rho = \\frac{(n_r + 2l + 1)!}{n_r!}\\,\\delta_{n_r n'_r}$$

*Nota pedagógica crucial:* En coordenadas esféricas, el elemento de volumen diferencial es $dV = r^2 dr \\sin\\theta d\\theta d\\phi$. Al integrar sobre la coordenada radial, el factor de volumen introduce un término $r^2 dr \\propto \\rho^2 d\\rho$. La función de onda radial total incluye el comportamiento asintótico $R \\propto \\rho^l e^{-\\rho/2}$, por lo que al evaluar la integral de superposición física:

$$\\int_0^\\infty R_{nl}(r) R_{n'l}(r)\\,r^2\\,dr \\propto \\int_0^\\infty L_{n-l-1}^{2l+1}(\\rho)\\,L_{n'-l-1}^{2l+1}(\\rho)\\,\\rho^{2l+2} e^{-\\rho}\\,d\\rho$$

La presencia de $\\rho^{2l+2}$ en la integral física frente a $\\rho^{2l+1}$ en la ecuación diferencial de Sturm-Liouville se resuelve recordando que la variable espacial $\\rho = 2\\kappa_n r$ escala con el autovalor de energía $\\kappa_n$, lo que conduce a las relaciones de recurrencia cruzadas de Laguerre.

**3. Completitud:**  
Las autofunciones ligadas forman una base ortogonal para cualquier estado ligado. Sin embargo, en el potencial de Coulomb, a diferencia del oscilador armónico, el espectro ligado discreto ($E < 0$) no cubre la totalidad del espacio de Hilbert $L^2(\\mathbb{R}^3)$; es imprescindible incluir el **continuo de estados de dispersión** ($E > 0$, donde el electrón escapa del núcleo) para obtener una base completa capaz de describir cualquier paquete de onda arbitrario.


```python
# ============================================================
# 🔬 Verificación Simbólica Formal: Operador de Sturm-Liouville
#    para los Polinomios Asociados de Laguerre con SymPy
# ============================================================
# Ecuación autoadjunta:
#   d/dρ [ ρ^{2l+2} · exp(-ρ) · dL/dρ ] + n_r · ρ^{2l+1} · exp(-ρ) · L = 0
# donde:
#   p(ρ) = ρ^{2l+2} · exp(-ρ)
#   w(ρ) = ρ^{2l+1} · exp(-ρ)
#   Λ    = n_r = n - l - 1
# ============================================================

rho_sym = smp.symbols('rho', positive=True)

def operador_SL_laguerre(y_expr, l_val, var=rho_sym):
    # Aplica el operador diferencial de Sturm-Liouville:
    # L[y] = - d/dρ [ p(ρ) * dy/dρ ] donde p(ρ) = ρ^{2l+2} * exp(-ρ)
    p_expr = (var**(2 * l_val + 2)) * smp.exp(-var)
    dy_dvar = smp.diff(y_expr, var)
    termino_flujo = p_expr * dy_dvar
    return -smp.diff(termino_flujo, var)

print("=" * 85)
print("  COMPROBACIÓN SIMBÓLICA CON SYMPY: L[L_{nr}^{2l+1}] - nr · w(ρ) · L == 0")
print("=" * 85)

casos_cuanticos = [
    # (n, l) -> nr = n - l - 1, alpha = 2l + 1
    (1, 0),  # Estado 1s: nr = 0, l = 0, alpha = 1
    (2, 0),  # Estado 2s: nr = 1, l = 0, alpha = 1
    (2, 1),  # Estado 2p: nr = 0, l = 1, alpha = 3
    (3, 0),  # Estado 3s: nr = 2, l = 0, alpha = 1
    (3, 1),  # Estado 3p: nr = 1, l = 1, alpha = 3
    (3, 2),  # Estado 3d: nr = 0, l = 2, alpha = 5
]

for n_p, l_p in casos_cuanticos:
    nr_val = n_p - l_p - 1
    alpha_val = 2 * l_p + 1
    
    # Polinomio asociado de Laguerre simbólico exacto
    L_poly = smp.assoc_laguerre(nr_val, alpha_val, rho_sym)
    
    # Operador diferencial
    L_res = operador_SL_laguerre(L_poly, l_p, var=rho_sym)
    
    # Lado derecho: Λ * w(ρ) * L
    w_expr = (rho_sym**(alpha_val)) * smp.exp(-rho_sym)
    lado_der = nr_val * w_expr * L_poly
    
    residuo = smp.simplify(L_res - lado_der)
    
    estado_str = f"n={n_p}, l={l_p} (nr={nr_val})"
    print(f"Estado {estado_str:<18} | L_{nr_val}^{alpha_val}(ρ) = {str(L_poly):<22}")
    print(f"   Operador L[L]     = {smp.factor(L_res)}")
    print(f"   Término Λ·w(ρ)·L  = {smp.factor(lado_der)}")
    print(f"   Residuo analítico = {residuo}  ✓ (¡Idénticamente nulo!)")
    print("-" * 85)

print("🎉 Todos los Polinomios Asociados de Laguerre satisfacen con exactitud analítica")
print("   la ecuación canónica autoadjunta de Sturm-Liouville.")
```

```python
# ============================================================
# 📊 Funciones Radiales R_{nl}(r) y Densidad de Probabilidad Radial
# ============================================================
# R_{nl}(r) = N_{nl} · ρ^l · exp(-ρ/2) · L_{n-l-1}^{2l+1}(ρ)
# Densidad por capa esférica: P(r) = r² |R_{nl}(r)|²
# ============================================================

def R_radial_hidrogeno(n, l, r_vals, a0=1.0):
    # Calcula la función de onda radial R_{nl}(r) para el átomo hidrogenoide
    nr = n - l - 1
    alpha = 2 * l + 1
    rho = 2.0 * r_vals / (n * a0)
    
    norma = np.sqrt(
        (2.0 / (n * a0))**3 * (math.factorial(nr) / (2.0 * n * math.factorial(n + l)))
    )
    L_val = sp.eval_genlaguerre(nr, alpha, rho)
    return norma * (rho**l) * np.exp(-rho / 2.0) * L_val

r_grid = np.linspace(0.001, 24.0, 600)

fig, axes = plt.subplots(1, 3, figsize=(17, 5))
fig.patch.set_facecolor('#0d1117')

# ------------------------------------------------------------
# 1. Funciones de Onda Radiales R_{nl}(r)
# ------------------------------------------------------------
ax1 = axes[0]
estados_graf = [(1, 0, '1s'), (2, 0, '2s'), (2, 1, '2p'), (3, 0, '3s'), (3, 2, '3d')]
colores_orb = ['#58a6ff', '#3fb950', '#d29922', '#f85149', '#bc8cff']

for (n, l, nombre), col in zip(estados_graf, colores_orb):
    R_vals = R_radial_hidrogeno(n, l, r_grid)
    ax1.plot(r_grid, R_vals, label=rf'${nombre}$ ($n={n}, l={l}$)', color=col, lw=2.2)

ax1.axhline(0, color='white', lw=0.5, alpha=0.3)
ax1.set_title(r'1. Funciones Radiales $R_{nl}(r)$', fontweight='bold')
ax1.set_xlabel(r'Radio $r / a_0$'); ax1.set_ylabel(r'$R_{nl}(r)$')
ax1.set_xlim(0, 20); ax1.legend(fontsize=9, framealpha=0.3); ax1.grid(True)

# ------------------------------------------------------------
# 2. Densidad de Probabilidad Radial P(r) = r² |R_{nl}(r)|²
# ------------------------------------------------------------
ax2 = axes[1]
for (n, l, nombre), col in zip(estados_graf, colores_orb):
    R_vals = R_radial_hidrogeno(n, l, r_grid)
    P_radial = (r_grid**2) * (R_vals**2)
    ax2.plot(r_grid, P_radial, label=rf'${nombre}$', color=col, lw=2.2)
    ax2.fill_between(r_grid, 0, P_radial, color=col, alpha=0.15)

ax2.set_title(r'2. Densidad de Probabilidad Radial $P(r) = r^2 |R_{nl}|^2$', fontweight='bold')
ax2.set_xlabel(r'Radio $r / a_0$'); ax2.set_ylabel(r'$P(r)$')
ax2.set_xlim(0, 22); ax2.legend(fontsize=9, framealpha=0.3); ax2.grid(True)

# ------------------------------------------------------------
# 3. Función de peso de Sturm-Liouville w(ρ) = ρ^{2l+1} exp(-ρ)
# ------------------------------------------------------------
ax3 = axes[2]
rho_plot = np.linspace(0.01, 14.0, 400)
for l_idx, col in zip([0, 1, 2], ['#58a6ff', '#3fb950', '#d29922']):
    alpha_idx = 2 * l_idx + 1
    w_rho = (rho_plot**alpha_idx) * np.exp(-rho_plot)
    ax3.plot(rho_plot, w_rho, label=rf'$l={l_idx}$ ($w(\rho)=\rho^{{{alpha_idx}}} e^{{-\rho}}$)',
             color=col, lw=2.2)
    ax3.fill_between(rho_plot, 0, w_rho, color=col, alpha=0.15)

ax3.set_title(r'3. Función de Peso S-L $w(\rho) = \rho^{2l+1} e^{-\rho}$', fontweight='bold')
ax3.set_xlabel(r'$\rho = 2\kappa r$'); ax3.set_ylabel(r'$w(\rho)$')
ax3.set_xlim(0, 14); ax3.legend(fontsize=9, framealpha=0.3); ax3.grid(True)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 🔬 Matriz de Gram para Polinomios Asociados de Laguerre
# ============================================================
# Teorema S-L:
#   ∫_0^∞ L_p^α(ρ) L_q^α(ρ) ρ^α e^{-ρ} dρ = [(p + α)! / p!] · δ_{pq}
# Fijamos l = 1 (orbitales tipo p, α = 2l + 1 = 3)
# ============================================================

l_fijo = 1
alpha_fijo = 2 * l_fijo + 1  # α = 3
N_radial = 4  # nr = 0, 1, 2, 3
G_laguerre = np.zeros((N_radial, N_radial))

print("=" * 85)
print(f"  MATRIZ DE GRAM PONDERADA PARA LAGUERRE CON l={l_fijo} (α={alpha_fijo})")
print("=" * 85)
print(f"{'Par (p, q)':<12} | {'Exacto: (p + α)! / p!':<25} | {'Numérico (SciPy quad)':<25} | {'Error':<15}")
print("-" * 85)

for p in range(N_radial):
    for q in range(N_radial):
        val_num, _ = integrate.quad(
            lambda r: sp.eval_genlaguerre(p, alpha_fijo, r) * 
                      sp.eval_genlaguerre(q, alpha_fijo, r) * 
                      (r**alpha_fijo) * np.exp(-r),
            0, np.inf
        )
        G_laguerre[p, q] = val_num
        
        val_exacto = float(math.factorial(p + alpha_fijo) / math.factorial(p)) if p == q else 0.0
        error = abs(val_num - val_exacto)
        
        if p == q or p < q:
            print(f"({p}, {q}){' ':>6} | {val_exacto:<25.1f} | {val_num:<25.6f} | {error:<15.2e}")

print("=" * 85)

fig, ax = plt.subplots(figsize=(6.5, 5))
fig.patch.set_facecolor('#0d1117')
cax = ax.imshow(G_laguerre, cmap='viridis', interpolation='nearest')
ax.set_title(rf'Matriz de Gram: Laguerre $\langle L_p^{{3}}, L_q^{{3}} \rangle_{{w(\rho)}}$', fontweight='bold')
ax.set_xticks(range(N_radial)); ax.set_xticklabels([f'$L_{k}^{{3}}$' for k in range(N_radial)])
ax.set_yticks(range(N_radial)); ax.set_yticklabels([f'$L_{k}^{{3}}$' for k in range(N_radial)])
fig.colorbar(cax, ax=ax, fraction=0.046, pad=0.04)

for r in range(N_radial):
    for c in range(N_radial):
        val = G_laguerre[r, c]
        color_txt = 'black' if val > G_laguerre.max()*0.5 else 'white'
        ax.text(c, r, f'{val:.1f}', ha='center', va='center',
                color=color_txt, fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 🌐 Visualización 3D Interactiva con k3d:
#    Isosuperficies de Orbitales Atómicos (Marching Cubes)
# ============================================================
# Renderizamos las isosuperficies de densidad de probabilidad |ψ|²
# coloreando por el signo o fase de la función de onda:
#   - Lóbulo positivo: Azul / Cian
#   - Lóbulo negativo: Rojo / Coral
# ============================================================

def orbital_atomico_3d(estado='2p_z', N_grid=55, r_max=12.0):
    # Calcula el campo escalar 3D del orbital atómico en coordenadas cartesianas
    lin = np.linspace(-r_max, r_max, N_grid)
    Z_mesh, Y_mesh, X_mesh = np.meshgrid(lin, lin, lin, indexing='ij')
    R_mesh = np.sqrt(X_mesh**2 + Y_mesh**2 + Z_mesh**2) + 1e-12
    a0 = 1.0
    
    if estado == '1s':
        psi = (1.0 / np.sqrt(np.pi * a0**3)) * np.exp(-R_mesh / a0)
        iso_sug = 0.008
    elif estado == '2s':
        psi = (1.0 / (4.0 * np.sqrt(2.0 * np.pi * a0**3))) * (2.0 - R_mesh / a0) * np.exp(-R_mesh / (2.0 * a0))
        iso_sug = 0.001
    elif estado == '2p_z':
        cos_theta = Z_mesh / R_mesh
        psi = (1.0 / (4.0 * np.sqrt(2.0 * np.pi * a0**3))) * (R_mesh / a0) * np.exp(-R_mesh / (2.0 * a0)) * cos_theta
        iso_sug = 0.0008
    elif estado == '2p_x':
        sin_cos = X_mesh / R_mesh
        psi = (1.0 / (4.0 * np.sqrt(2.0 * np.pi * a0**3))) * (R_mesh / a0) * np.exp(-R_mesh / (2.0 * a0)) * sin_cos
        iso_sug = 0.0008
    elif estado == '3d_z2':
        # 3z² - r²
        factor_angular = (3.0 * Z_mesh**2 - R_mesh**2) / (R_mesh**2)
        psi = (1.0 / (81.0 * np.sqrt(6.0 * np.pi * a0**3))) * (R_mesh / a0)**2 * np.exp(-R_mesh / (3.0 * a0)) * factor_angular
        iso_sug = 0.0001
    elif estado == '3d_xy':
        factor_angular = (X_mesh * Y_mesh) / (R_mesh**2)
        psi = (1.0 / (81.0 * np.sqrt(2.0 * np.pi * a0**3))) * (R_mesh / a0)**2 * np.exp(-R_mesh / (3.0 * a0)) * factor_angular
        iso_sug = 0.0001
    else:  # 3s
        rho_3 = 2.0 * R_mesh / (3.0 * a0)
        L_3s = sp.eval_genlaguerre(2, 1, rho_3)
        psi = (2.0 / (81.0 * np.sqrt(3.0 * a0**3))) * L_3s * np.exp(-rho_3 / 2.0)
        iso_sug = 0.0003
        
    prob = (psi**2).astype(np.float32)
    return prob, psi.astype(np.float32), iso_sug, lin

# Configuración del plot k3d
plot_orbital = k3d.plot(grid_auto_fit=True, camera_auto_fit=True, height=600)

estado_inicial = '2p_z'
prob_init, psi_init, iso_init, lin_init = orbital_atomico_3d(estado_inicial, N_grid=55, r_max=12.0)
lim_box = float(lin_init[-1])

max_psi = float(np.max(np.abs(psi_init)))
mc_orbital = k3d.marching_cubes(
    prob_init,
    level=iso_init,
    attribute=psi_init,
    color_map=matplotlib_color_maps.Coolwarm,
    color_range=[-max_psi, max_psi],
    xmin=-lim_box, xmax=lim_box,
    ymin=-lim_box, ymax=lim_box,
    zmin=-lim_box, zmax=lim_box,
    flat_shading=False
)
plot_orbital += mc_orbital

# Controles interactivos
drop_orbital = widgets.Dropdown(
    options=['1s', '2s', '2p_z', '2p_x', '3s', '3d_z2', '3d_xy'],
    value=estado_inicial,
    description='Orbital:'
)
slider_iso = widgets.FloatLogSlider(
    value=iso_init, base=10, min=-5, max=-1, step=0.1,
    description='Iso-nivel:',
    layout=widgets.Layout(width='400px')
)

def actualizar_orbital(change):
    est = drop_orbital.value
    prob_n, psi_n, iso_sug, _ = orbital_atomico_3d(est, N_grid=55, r_max=12.0)
    
    # Actualizar campos
    mc_orbital.scalar_field = prob_n
    mc_orbital.attribute = psi_n
    max_p = float(np.max(np.abs(psi_n)))
    mc_orbital.color_range = [-max_p, max_p]
    
    # Ajustar nivel
    if change and change.get('owner') == drop_orbital:
        slider_iso.value = iso_sug
    mc_orbital.level = slider_iso.value

drop_orbital.observe(actualizar_orbital, names='value')
slider_iso.observe(actualizar_orbital, names='value')

display(widgets.HBox([drop_orbital, slider_iso]))
plot_orbital.display()
print(f"🪐 Visualizando orbital {estado_inicial}")
print("   → Cambia de orbital o ajusta el slider de iso-nivel para ver los nodos esféricos/cónicos.")
```

```python
# ============================================================
# 🎬 Animación 3D Interactiva con k3d:
#    Transición Cuántica 1s ↔ 2p_z (Dipolo Radiante de Bohr)
# ============================================================
# Superposición cuántica entre el estado fundamental y el primer excitado:
#   Ψ(r, t) = [ψ₁₀₀ + ψ₂₁₀ e^{-i ω₂₁ t}] / √2
# La nube electrónica oscila físicamente a lo largo del eje z
# con frecuencia angular ω₂₁ = (E₂ - E₁) / ℏ.
# ¡Este es el origen cuántico de la radiación electromagnética!
# ============================================================

plot_anim_trans = k3d.plot(grid_auto_fit=False, camera_auto_fit=False, height=600)

N_grid_trans = 45
r_box_trans = 9.0
lin_t = np.linspace(-r_box_trans, r_box_trans, N_grid_trans)
Z_t, Y_t, X_t = np.meshgrid(lin_t, lin_t, lin_t, indexing='ij')
R_t = np.sqrt(X_t**2 + Y_t**2 + Z_t**2) + 1e-12
a0 = 1.0

# Funciones de onda 1s y 2pz
u_1s = (1.0 / np.sqrt(np.pi * a0**3)) * np.exp(-R_t / a0)
u_2pz = (1.0 / (4.0 * np.sqrt(2.0 * np.pi * a0**3))) * (R_t / a0) * np.exp(-R_t / (2.0 * a0)) * (Z_t / R_t)

N_frames_trans = 40
fases_trans = np.linspace(0, 2*np.pi, N_frames_trans)

# Cuadro inicial t = 0
dens_trans_0 = 0.5 * (u_1s + u_2pz)**2
mc_trans = k3d.marching_cubes(
    dens_trans_0.astype(np.float32),
    level=0.0025,
    color=0x58a6ff,
    xmin=-r_box_trans, xmax=r_box_trans,
    ymin=-r_box_trans, ymax=r_box_trans,
    zmin=-r_box_trans, zmax=r_box_trans,
    flat_shading=False
)
plot_anim_trans += mc_trans

play_trans = widgets.Play(
    value=0, min=0, max=N_frames_trans - 1, step=1,
    interval=60, description="Play"
)
slider_trans = widgets.IntSlider(
    min=0, max=N_frames_trans - 1, step=1, description='Fase ω₂₁t:',
    layout=widgets.Layout(width='400px')
)
widgets.jslink((play_trans, 'value'), (slider_trans, 'value'))

def actualizar_dipolo(change):
    fase = fases_trans[slider_trans.value]
    # |u_1s + u_2pz e^{-iωt}|² = u_1s² + u_2pz² + 2 u_1s u_2pz cos(ωt)
    dens_actual = 0.5 * (u_1s**2 + u_2pz**2 + 2.0 * u_1s * u_2pz * np.cos(fase))
    mc_trans.scalar_field = dens_actual.astype(np.float32)

slider_trans.observe(actualizar_dipolo, names='value')

display(widgets.HBox([play_trans, slider_trans]))
plot_anim_trans.display()
print("💡 ¡Presiona ▶ Play para observar la oscilación del dipolo atómico!")
print("   La densidad de probabilidad se desplaza rítmicamente entre +z y -z,")
print("   generando una corriente dipolar que emite un fotón de energía ℏω₂₁ = 10.2 eV.")
```

### 🎓 Lo que aprendimos

1. **Unificación por Sturm–Liouville:**  
   Tanto el oscilador armónico cuántico como el átomo de hidrógeno son **problemas singulares de Sturm-Liouville**. Al extraer el comportamiento asintótico adecuado ($e^{-\\xi^2/2}$ en el OAC y $\\rho^l e^{-\\rho/2}$ en el hidrógeno), las ecuaciones diferenciales de Schrödinger se reducen a formas polinomiales canónicas (Hermite y Laguerre).
   
2. **Origen Matemático de la Cuantización:**  
   La cuantización cuántica de la energía **no es un postulado misterioso**, sino la condición analítica necesaria de la teoría de Sturm-Liouville para que la serie de potencias se trunque y las autofunciones pertenezcan al espacio de Hilbert $L^2$.
   
3. **El Significado de la Función de Peso $w(x)$:**  
   La función de peso $w(\\xi) = e^{-\\xi^2}$ (OAC) y $w(\\rho) = \\rho^{2l+1} e^{-\\rho}$ (Hidrógeno) determina el producto interno ponderado bajo el cual las autofunciones son ortogonales, reflejando tanto la geometría del espacio (factores de escala en coordenadas esféricas) como el potencial confinante.
   
4. **Correspondencia Clásica y Transiciones Radiativas:**  
   Las superposiciones cuánticas de autofunciones explican fenómenos dinámicos fundamentales: los paquetes de onda oscilan como partículas clásicas en el pozo armónico, y la interferencia entre los estados $1s$ y $2p_z$ genera un dipolo eléctrico oscilante que emite radiación electromagnética (fotones).
