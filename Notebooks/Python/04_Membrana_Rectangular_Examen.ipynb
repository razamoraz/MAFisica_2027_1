## 0. Fenómeno Físico: Membrana Rectangular con Bordes Fijos

Consideremos una membrana elástica homogénea bidimensional que ocupa la región cuadrada unitaria:
$$\Omega = \left\{ (x,y) \in \mathbb{R}^2 : 0 \le x \le 1,\ 0 \le y \le 1 \right\}$$

Sus cuatro bordes perimetrales están rígidamente empotrados (desplazamiento vertical nulo). Sea $u(x,y,t)$ la desviación transversal respecto al plano de equilibrio en el punto $(x,y)$ e instante $t$.

Bajo la hipótesis de pequeñas deformaciones y tensión superficial constante $\tau$ con densidad de masa superficial $\sigma$, la velocidad de onda es $c = \sqrt{\tau/\sigma} > 0$. El fenómeno está gobernado por la **ecuación de onda bidimensional**:

$$u_{tt} = c^2 \left( u_{xx} + u_{yy} \right) = c^2 \nabla^2 u, \qquad (x,y) \in (0,1)^2,\ t > 0$$

con **condiciones de frontera de Dirichlet homogéneas**:
$$u(0,y,t) = u(1,y,t) = 0, \qquad u(x,0,t) = u(x,1,t) = 0 \qquad \forall t \ge 0$$

y **condiciones iniciales** impuestas:
- Desplazamiento inicial plano uniforme: $u(x,y,0) = 1$
- Velocidad inicial nula (liberada del reposo): $u_t(x,y,0) = 0$

Proponemos la solución mediante el **método de separación de variables**:
$$u(x,y,t) = X(x)\,Y(y)\,T(t)$$

## Pregunta 1 — Separación de variables y problemas de Sturm–Liouville (10 pts)

### 1.1 Sustitución en la ecuación de onda y división entre $XYT$
Calculamos las derivadas parciales de la propuesta $u(x,y,t) = X(x)Y(y)T(t)$:
- Derivada temporal: $u_{tt} = X(x)Y(y)T''(t)$
- Derivadas espaciales: $u_{xx} = X''(x)Y(y)T(t)$, \quad $u_{yy} = X(x)Y''(y)T(t)$

Sustituyendo en la ecuación de onda:
$$X(x)Y(y)T''(t) = c^2 \left[ X''(x)Y(y)T(t) + X(x)Y''(y)T(t) \right]$$

Dividiendo toda la ecuación entre $c^2 X(x)Y(y)T(t)$ (para soluciones no triviales $u \neq 0$):
$$\boxed{\frac{T''}{c^2 T} = \frac{X''}{X} + \frac{Y''}{Y}}$$

---

### 1.2 Introducción de la constante de separación temporal $\lambda$
El miembro izquierdo depende **únicamente del tiempo $t$**, mientras que el miembro derecho depende **únicamente de las coordenadas espaciales $(x,y)$**:
$$\underbrace{\frac{T''(t)}{c^2 T(t)}}_{\text{función de } t} = \underbrace{\frac{X''(x)}{X(x)} + \frac{Y''(y)}{Y(y)}}_{\text{función de } (x,y)}$$

Para que esta igualdad se verifique para todo $t > 0$ y todo $(x,y) \in (0,1)^2$, ambos miembros deben ser iguales a una **constante de separación** idéntica. Denotemos esta constante por $-\lambda$ (donde elegimos el signo negativo para asegurar soluciones temporales acotadas y periódicas / oscilatorias):
$$\frac{T''}{c^2 T} = -\lambda$$

Multiplicando por $c^2 T$, obtenemos la ecuación diferencial ordinaria para la parte temporal:
$$\boxed{T''(t) + c^2\lambda\,T(t) = 0} \quad \Longleftrightarrow \quad T''(t) + \omega^2 T(t) = 0, \quad \text{con } \omega = c\sqrt{\lambda}$$

---

### 1.3 Separación de la parte espacial: constantes $\mu$, $\nu$ y relación $\mu + \nu = \lambda$
Tomamos ahora la ecuación espacial resultante:
$$\frac{X''(x)}{X(x)} + \frac{Y''(y)}{Y(y)} = -\lambda$$

Aislamos los términos que dependen de $x$ y los que dependen de $y$:
$$\frac{X''(x)}{X(x)} = -\lambda - \frac{Y''(y)}{Y(y)}$$

Nuevamente, el lado izquierdo depende exclusivamente de $x$ y el lado derecho depende exclusivamente de $y$. Por tanto, ambos deben igualar a una nueva constante independiente, que denotamos $-\mu$:
$$\frac{X''}{X} = -\mu \implies X'' + \mu X = 0$$

Sustituyendo de regreso en la ecuación espacial:
$$-\mu + \frac{Y''}{Y} = -\lambda \implies \frac{Y''}{Y} = -(\lambda - \mu) \equiv -\nu$$

Definiendo $\nu = \lambda - \mu$, obtenemos:
$$\frac{Y''}{Y} = -\nu \implies Y'' + \nu Y = 0$$

con la relación fundamental de compatibilidad:
$$\boxed{\mu + \nu = \lambda}$$

---

### 1.4 Formulación formal de los dos problemas de Sturm–Liouville independientes
Las condiciones de frontera para la membrana exigen:
- En $x = 0$ y $x = 1$: $u(0,y,t) = X(0)Y(y)T(t) = 0$ y $u(1,y,t) = X(1)Y(y)T(t) = 0$. Para soluciones no triviales ($Y(y)T(t) \not\equiv 0$), esto implica $X(0) = X(1) = 0$.
- En $y = 0$ e $y = 1$: $u(x,0,t) = X(x)Y(0)T(t) = 0$ y $u(x,1,t) = X(x)Y(1)T(t) = 0$. Para soluciones no triviales, $Y(0) = Y(1) = 0$.

Recordando la forma canónica del operador de Sturm–Liouville regular:
$$\mathcal{L}[y] \equiv -\frac{d}{dx}\left[p(x)\frac{dy}{dx}\right] + q(x)y = \lambda_0 w(x)y$$
con coeficientes estándar $p(x) = 1$, $q(x) = 0$, $w(x) = 1$, los dos problemas independientes quedan formalmente planteados como:

$$\boxed{\begin{aligned}
\textbf{Problema de Sturm–Liouville para } X(x): & \quad \begin{cases} -X''(x) = \mu X(x), & 0 < x < 1 \\ X(0) = 0, \quad X(1) = 0 & \end{cases} \\ \\ 
\textbf{Problema de Sturm–Liouville para } Y(y): & \quad \begin{cases} -Y''(y) = \nu Y(y), & 0 < y < 1 \\ Y(0) = 0, \quad Y(1) = 0 & \end{cases}
\end{aligned}}$$


```python
# ============================================================
# 🔬 Verificación Simbólica con SymPy — Pregunta 1
# ============================================================
import sympy as sp

# Símbolos matemáticos
x, y, t, c = sp.symbols('x y t c', positive=True)
mu, nu, lam = sp.symbols('mu nu lambda', positive=True)
X = sp.Function('X')(x)
Y = sp.Function('Y')(y)
T = sp.Function('T')(t)

u = X * Y * T

# Operadores de la ecuación de onda
u_tt = sp.diff(u, t, 2)
u_xx = sp.diff(u, x, 2)
u_yy = sp.diff(u, y, 2)

wave_eq = sp.Eq(u_tt, c**2 * (u_xx + u_yy))
print("Ecuación de onda 2D:")
sp.pprint(wave_eq)

# División formal entre c^2 * X * Y * T
separada = sp.simplify((u_tt / (c**2 * u)) - ((u_xx + u_yy) / (X * Y)))
print("\nComprobación de identidad de separación (debe ser 0):", separada)

```

## Pregunta 2 — Autovalores, autofunciones y frecuencias (10 pts)

### 2.1 Solución general de $-X'' = \mu X$ con $X(0) = X(1) = 0$
Analizamos el problema diferencial:
$$X''(x) + \mu X(x) = 0$$

- **Caso $\mu < 0$:** Sea $\mu = -\kappa^2$ con $\kappa > 0$. La solución general es $X(x) = C_1 \cosh(\kappa x) + C_2 \sinh(\kappa x)$.
  - $X(0) = C_1 = 0 \implies X(x) = C_2 \sinh(\kappa x)$.
  - $X(1) = C_2 \sinh(\kappa) = 0$. Como $\kappa > 0$, $\sinh(\kappa) \neq 0$, lo que obliga a $C_2 = 0$ (solución trivial).
- **Caso $\mu = 0$:** $X''(x) = 0 \implies X(x) = C_1 x + C_2$.
  - $X(0) = C_2 = 0 \implies X(1) = C_1 = 0$ (solución trivial).
- **Caso $\mu > 0$:** Definimos $\mu = k_x^2$ con $k_x = \sqrt{\mu} > 0$.
  La ecuación característica asociada es $r^2 + \mu = 0 \implies r = \pm i\sqrt{\mu}$.
  Por consiguiente, la **solución general** es:
  $$\boxed{X(x) = C_1 \cos(\sqrt{\mu}\,x) + C_2 \sin(\sqrt{\mu}\,x)}$$

---

### 2.2 Aplicación de condiciones de frontera y autovalores $\mu_m = m^2\pi^2$
Aplicamos las condiciones de frontera de Dirichlet homogéneas:
1. $X(0) = 0$:
   $$X(0) = C_1 \cos(0) + C_2 \sin(0) = C_1 = 0$$
   De este modo, la solución se reduce a:
   $$X(x) = C_2 \sin(\sqrt{\mu}\,x)$$

2. $X(1) = 0$:
   $$X(1) = C_2 \sin(\sqrt{\mu}) = 0$$
   Para obtener autofunciones no triviales, requerimos $C_2 \neq 0$, lo que conduce a la condición de cuantización:
   $$\sin(\sqrt{\mu}) = 0 \implies \sqrt{\mu} = m\pi, \qquad m \in \mathbb{Z}^+ = \{1, 2, 3, \dots\}$$

Elevando al cuadrado, demostramos que el espectro de autovalores es estrictamente discreto y positivo:
$$\boxed{\mu_m = m^2\pi^2, \qquad m = 1, 2, 3, \dots}$$

---

### 2.3 Autofunciones resultantes $X_m(x)$
Eligiendo la constante arbitraria de normalización como $C_2 = 1$, las autofunciones espaciales en la dirección $x$ son:
$$\boxed{X_m(x) = \sin(m\pi x), \qquad m = 1, 2, 3, \dots}$$

---

### 2.4 Autovalores y autofunciones para $Y(y)$
El problema de Sturm–Liouville para $Y(y)$:
$$-Y''(y) = \nu Y(y), \quad 0 < y < 1, \qquad Y(0) = Y(1) = 0$$
es matemáticamente idéntico al problema de $X(x)$ bajo la biyección $x \leftrightarrow y$. Por tanto, reportamos directamente:
$$\boxed{\nu_n = n^2\pi^2, \qquad Y_n(y) = \sin(n\pi y), \qquad n = 1, 2, 3, \dots}$$

---

### 2.5 Autovalor total $\lambda_{mn}$, frecuencia angular $\omega_{mn}$ y autofunción espacial conjunta $\phi_{mn}(x,y)$
Usando la relación de acoplamiento espacial $\lambda = \mu + \nu$:
- **Autovalor total conjunto:**
  $$\boxed{\lambda_{mn} = \mu_m + \nu_n = (m^2 + n^2)\pi^2, \qquad m, n = 1, 2, 3, \dots}$$

- **Frecuencia angular temporal:**
  $$\boxed{\omega_{mn} = c\sqrt{\lambda_{mn}} = c\pi\sqrt{m^2 + n^2}}$$

- **Autofunción espacial conjunta no normalizada:**
  $$\boxed{\phi_{mn}(x,y) = X_m(x)Y_n(y) = \sin(m\pi x)\sin(n\pi y)}$$

> 💡 **Degeneración modal:** Para una membrana cuadrada ($L_x = L_y = 1$), los modos con $m \neq n$ presentan **degeneración espectral bidimensional**: $\omega_{mn} = \omega_{nm}$ (por ejemplo, el modo $(1,2)$ y el modo $(2,1)$ oscilan exactamente a la misma frecuencia $\omega = \sqrt{5}\pi c$).


```python
# ============================================================
# 🎨 Visualización 2D — FIGURA 1: Atlas de Modos Normales y Líneas Nodales
# ============================================================
import numpy as np
import matplotlib.pyplot as plt

# Configuración de tema oscuro idéntica a 01_Bessel_Drums.ipynb
plt.style.use('dark_background')
plt.rcParams.update({
    'font.size': 11,
    'axes.titlesize': 12,
    'figure.facecolor': '#0d1117',
    'axes.facecolor': '#0d1117',
    'savefig.facecolor': '#0d1117',
})

# Malla espacial 2D en el dominio [0, 1] x [0, 1]
N = 250
x = np.linspace(0, 1, N)
y = np.linspace(0, 1, N)
X, Y = np.meshgrid(x, y)

# Selección de modos representativos
modos = [(1, 1), (1, 2), (2, 1), (2, 2), (3, 1), (3, 3)]
titulos = [
    r'Modo $(1,1)$ — Fundamental ($\omega_{11} = \sqrt{2}\pi c$)',
    r'Modo $(1,2)$ — $\omega_{12} = \sqrt{5}\pi c$',
    r'Modo $(2,1)$ — $\omega_{21} = \sqrt{5}\pi c$',
    r'Modo $(2,2)$ — $\omega_{22} = \sqrt{8}\pi c$',
    r'Modo $(3,1)$ — $\omega_{31} = \sqrt{10}\pi c$',
    r'Modo $(3,3)$ — $\omega_{33} = \sqrt{18}\pi c$'
]

fig, axes = plt.subplots(2, 3, figsize=(15, 9.5))
fig.patch.set_facecolor('#0d1117')

for ax, (m, n), tit in zip(axes.flat, modos, titulos):
    ax.set_facecolor('#0d1117')
    phi_mn = np.sin(m * np.pi * X) * np.sin(n * np.pi * Y)
    
    # Mapa de calor de amplitudes relativas
    im = ax.contourf(X, Y, phi_mn, levels=45, cmap='coolwarm', vmin=-1.0, vmax=1.0)
    ax.contour(X, Y, phi_mn, levels=8, colors='white', linewidths=0.4, alpha=0.3)
    
    # Trazado de líneas nodales interiores (donde phi_mn = 0)
    for j in range(1, m):
        ax.axvline(x=j/m, color='#58a6ff', linestyle='--', linewidth=1.8, label='Línea nodal' if j==1 and n==1 else '')
    for k in range(1, n):
        ax.axhline(y=k/n, color='#f97583', linestyle='--', linewidth=1.8)
        
    ax.set_title(tit, color='white', fontweight='bold', pad=8)
    ax.set_xlabel(r'$x$')
    ax.set_ylabel(r'$y$')
    ax.set_aspect('equal')
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)

# Barra de color común
cbar_ax = fig.add_axes([0.92, 0.15, 0.02, 0.7])
cbar = fig.colorbar(im, cax=cbar_ax)
cbar.set_label(r'Amplitud modal $\phi_{mn}(x,y)$', color='white', labelpad=10)
cbar.ax.tick_params(colors='white')

fig.suptitle(r'Atlas 2D de Autofunciones $\phi_{mn}(x,y) = \sin(m\pi x)\sin(n\pi y)$ y Cuadrículas Nodales',
             fontsize=16, color='white', fontweight='bold', y=0.98)
plt.tight_layout(rect=[0, 0, 0.9, 0.95])
plt.show()

print("💡 Cada modo (m,n) divide la membrana en m x n celdas rectangulares.")
print("   Las líneas discontinuas azules y rosas indican los nodos donde la membrana permanece en reposo.")

```

## Pregunta 3 — Normalización y ortogonalidad (10 pts)

El producto interno en el espacio de Hilbert $L^2\big([0,1] \times [0,1]\big)$ está definido por:
$$\langle f, g \rangle = \int_0^1 \int_0^1 f(x,y)\,g^*(x,y)\,dx\,dy$$

Utilizaremos la identidad de ortogonalidad clásica del análisis de Fourier:
$$\int_0^1 \sin^2(m\pi x)\,dx = \int_0^1 \frac{1 - \cos(2m\pi x)}{2}\,dx = \left[ \frac{x}{2} - \frac{\sin(2m\pi x)}{4m\pi} \right]_0^1 = \frac{1}{2}, \qquad \forall m \in \mathbb{Z}^+$$

---

### 3.1 Cálculo explícito de la norma al cuadrado $\Vert\phi_{mn}\Vert^2$
Por el Teorema de Fubini, como las variables espaciales están completamente desacopladas, la integral doble se factoriza de manera exacta en el producto de dos integrales unidimensionales independientes:
$$\Vert\phi_{mn}\Vert^2 = \langle \phi_{mn}, \phi_{mn} \rangle = \int_0^1 \int_0^1 \sin^2(m\pi x)\,\sin^2(n\pi y)\,dx\,dy$$

$$\Vert\phi_{mn}\Vert^2 = \left( \int_0^1 \sin^2(m\pi x)\,dx \right) \left( \int_0^1 \sin^2(n\pi y)\,dy \right)$$

Sustituyendo la integral dada $\int_0^1 \sin^2(k\pi s)\,ds = \frac{1}{2}$:
$$\Vert\phi_{mn}\Vert^2 = \left(\frac{1}{2}\right) \left(\frac{1}{2}\right) = \frac{1}{4}$$

Por lo tanto, la norma $L^2$ de cada autofunción conjunta es:
$$\boxed{\Vert\phi_{mn}\Vert = \sqrt{\frac{1}{4}} = \frac{1}{2}}$$

---

### 3.2 Construcción y expresión de la autofunción normalizada $\psi_{mn}(x,y)$
Normalizamos dividiendo la autofunción espacial $\phi_{mn}(x,y)$ entre su norma $\Vert\phi_{mn}\Vert$:
$$\psi_{mn}(x,y) = \frac{\phi_{mn}(x,y)}{\Vert\phi_{mn}\Vert} = \frac{\sin(m\pi x)\,\sin(n\pi y)}{1/2}$$

$$\boxed{\psi_{mn}(x,y) = 2\sin(m\pi x)\sin(n\pi y)}$$

---

### 3.3 Relación matemática de ortonormalidad usando deltas de Kronecker
Evaluamos el producto interno entre dos autofunciones normalizadas arbitrarias $\psi_{mn}$ y $\psi_{pq}$:
$$\langle \psi_{mn}, \psi_{pq} \rangle = \int_0^1 \int_0^1 \left[2\sin(m\pi x)\sin(n\pi y)\right] \left[2\sin(p\pi x)\sin(q\pi y)\right] dx\,dy$$

Factorizando por Fubini:
$$\langle \psi_{mn}, \psi_{pq} \rangle = 4 \left( \int_0^1 \sin(m\pi x)\sin(p\pi x)\,dx \right) \left( \int_0^1 \sin(n\pi y)\sin(q\pi y)\,dy \right)$$

Recordando la relación de ortogonalidad fundamental de la base de Fourier seno:
$$\int_0^1 \sin(m\pi x)\sin(p\pi x)\,dx = \frac{1}{2}\delta_{mp}, \qquad \int_0^1 \sin(n\pi y)\sin(q\pi y)\,dy = \frac{1}{2}\delta_{nq}$$

Sustituyendo directamente:
$$\langle \psi_{mn}, \psi_{pq} \rangle = 4 \left( \frac{1}{2}\delta_{mp} \right) \left( \frac{1}{2}\delta_{nq} \right) = \delta_{mp}\,\delta_{nq}$$

$$\boxed{\langle \psi_{mn}, \psi_{pq} \rangle = \delta_{mp}\,\delta_{nq} = \begin{cases} 1, & \text{si } m = p \text{ y } n = q \\ 0, & \text{en cualquier otro caso} \end{cases}}$$


```python
# ============================================================
# 🔬 Verificación Numérica: Matriz de Gram de Ortonormalidad
# ============================================================
from scipy.integrate import dblquad

def psi(m, n, x, y):
    return 2.0 * np.sin(m * np.pi * x) * np.sin(n * np.pi * y)

test_modes = [(1, 1), (1, 2), (2, 1), (2, 2), (3, 1)]
gram_matrix = np.zeros((len(test_modes), len(test_modes)))

print("=" * 68)
print("  VERIFICACIÓN COMPUTACIONAL DE ORTONORMALIDAD: ⟨ψ_{mn}, ψ_{pq}⟩")
print("=" * 68)

for i, (m, n) in enumerate(test_modes):
    for j, (p, q) in enumerate(test_modes):
        val, _ = dblquad(lambda y, x: psi(m, n, x, y) * psi(p, q, x, y), 0, 1, 0, 1)
        gram_matrix[i, j] = val

headers = [f"({m},{n})" for m, n in test_modes]
print(" Modo \ Modo | " + " | ".join(headers) + " |")
print("-" * 68)
for i, h in enumerate(headers):
    row_str = f"   {h:6s}   | " + " | ".join([f"{gram_matrix[i, j]:7.4f}" for j in range(len(test_modes))]) + " |"
    print(row_str)
print("=" * 68)
print("✅ La matriz de Gram es exactamente la identidad (δ_{mp} δ_{nq}).")

```

## Pregunta 4 — Expansión de la condición inicial (10 pts)

### 4.1 Solución general en serie doble e imposición de velocidad inicial nula
La solución temporal general para el modo $(m,n)$ satisface $T_{mn}'' + \omega_{mn}^2 T_{mn} = 0$:
$$T_{mn}(t) = A_{mn}\cos(\omega_{mn}t) + B_{mn}\sin(\omega_{mn}t)$$

Por el principio de superposición lineal, la solución general de la ecuación de onda que cumple las condiciones de frontera de Dirichlet homogéneas es la serie doble de Fourier:
$$u(x,y,t) = \sum_{m=1}^\infty \sum_{n=1}^\infty \left[ A_{mn}\cos(\omega_{mn}t) + B_{mn}\sin(\omega_{mn}t) \right] \psi_{mn}(x,y)$$

Derivando término a término respecto a $t$ para obtener la velocidad transversal:
$$u_t(x,y,t) = \sum_{m=1}^\infty \sum_{n=1}^\infty \omega_{mn} \left[ -A_{mn}\sin(\omega_{mn}t) + B_{mn}\cos(\omega_{mn}t) \right] \psi_{mn}(x,y)$$

Evaluando en $t = 0$:
$$u_t(x,y,0) = \sum_{m=1}^\infty \sum_{n=1}^\infty \omega_{mn} B_{mn}\,\psi_{mn}(x,y) = 0$$

Dado que las autofunciones $\{\psi_{mn}\}$ constituyen una base ortonormal completa de $L^2\big([0,1]^2\big)$ y las frecuencias angulares $\omega_{mn} = c\pi\sqrt{m^2+n^2} > 0$, la única forma de que la serie se anule idénticamente es que cada uno de sus coeficientes sea cero:
$$\omega_{mn} B_{mn} = 0 \implies \boxed{B_{mn} = 0, \qquad \forall m, n \ge 1}$$

Así, la solución queda expresada únicamente en términos de los modos coseno:
$$u(x,y,t) = \sum_{m=1}^\infty \sum_{n=1}^\infty A_{mn}\cos(\omega_{mn}t)\,\psi_{mn}(x,y)$$

---

### 4.2 Proyección ortogonal de la condición $u(x,y,0) = 1$
Evaluando en $t = 0$, la membrana adopta su configuración inicial plana:
$$u(x,y,0) = \sum_{m=1}^\infty \sum_{n=1}^\infty A_{mn}\,\psi_{mn}(x,y) = 1$$

Tomamos el producto interno formal con un modo arbitrario $\psi_{pq}(x,y)$:
$$\langle 1, \psi_{pq} \rangle = \left\langle \sum_{m=1}^\infty \sum_{n=1}^\infty A_{mn}\,\psi_{mn}, \; \psi_{pq} \right\rangle = \sum_{m=1}^\infty \sum_{n=1}^\infty A_{mn}\,\underbrace{\langle \psi_{mn}, \psi_{pq} \rangle}_{\delta_{mp}\delta_{nq}} = A_{pq}$$

Por lo tanto, la fórmula integral para el coeficiente de Fourier es:
$$A_{mn} = \langle 1, \psi_{mn} \rangle = \int_0^1 \int_0^1 1 \cdot \psi_{mn}(x,y)\,dx\,dy$$

Sustituyendo $\psi_{mn}(x,y) = 2\sin(m\pi x)\sin(n\pi y)$:
$$\boxed{A_{mn} = 2 \left( \int_0^1 \sin(m\pi x)\,dx \right) \left( \int_0^1 \sin(n\pi y)\,dy \right)}$$

---

### 4.3 Resolución de la integral unidimensional $\int_0^1 \sin(m\pi x)\,dx$
Calculamos la integral definida analíticamente:
$$\int_0^1 \sin(m\pi x)\,dx = \left[ -\frac{\cos(m\pi x)}{m\pi} \right]_0^1 = -\frac{\cos(m\pi) - \cos(0)}{m\pi} = \frac{1 - (-1)^m}{m\pi}$$

Evaluando según la paridad del entero $m \in \mathbb{Z}^+$:
$$\boxed{\int_0^1 \sin(m\pi x)\,dx = \begin{cases} 0, & \text{si } m \text{ es par} \\ \dfrac{2}{m\pi}, & \text{si } m \text{ es impar} \end{cases}}$$

---

### 4.4 Determinación de $A_{mn}$ en todos los casos de paridad
Sustituyendo el resultado unidimensional en la expresión de $A_{mn}$:
$$A_{mn} = 2 \left[ \frac{1 - (-1)^m}{m\pi} \right] \left[ \frac{1 - (-1)^n}{n\pi} \right] = \frac{2\,[1 - (-1)^m][1 - (-1)^n]}{\pi^2 m n}$$

Analizamos los dos casos exhaustivos:
1. **Si $m$ es par o $n$ es par:**
   Si $m$ es par, $(-1)^m = 1 \implies 1 - (-1)^m = 0$.
   Si $n$ es par, $(-1)^n = 1 \implies 1 - (-1)^n = 0$.
   En cualquier caso, el producto se anula:
   $$\boxed{A_{mn} = 0 \qquad (\text{si } m \text{ es par } \mathbf{o} \; n \text{ es par})}$$

2. **Si ambos $m$ y $n$ son impares:**
   Para $m, n \in \{1, 3, 5, \dots\}$, tenemos $(-1)^m = -1$ y $(-1)^n = -1$, de modo que:
   $$1 - (-1)^m = 2, \qquad 1 - (-1)^n = 2$$
   $$A_{mn} = 2 \left( \frac{2}{m\pi} \right) \left( \frac{2}{n\pi} \right) = \frac{8}{\pi^2 m n}$$
   $$\boxed{A_{mn} = \frac{8}{\pi^2 m n} \qquad (\text{para } m, n \text{ impares})}$$

Parametrizando con enteros no negativos $j, k \ge 1$ mediante $m = 2j - 1$ y $n = 2k - 1$:
$$A_{2j-1, 2k-1} = \frac{8}{\pi^2 (2j-1)(2k-1)}$$

---

### 4.5 Serie de Fourier resultante para la condición inicial $u_0(x,y) = 1$
Expresando el estado inicial $1$ en términos de la base ortonormal $\{\psi_{mn}\}$:
$$\boxed{1 = \sum_{j=1}^\infty \sum_{k=1}^\infty \frac{8}{\pi^2 (2j-1)(2k-1)}\,\psi_{2j-1, 2k-1}(x,y)}$$

O bien, sustituyendo la autofunción $\psi_{mn}(x,y) = 2\sin(m\pi x)\sin(n\pi y)$:
$$\boxed{1 = \frac{16}{\pi^2} \sum_{j=1}^\infty \sum_{k=1}^\infty \frac{\sin\big((2j-1)\pi x\big)\,\sin\big((2k-1)\pi y\big)}{(2j-1)(2k-1)}}$$

## Pregunta 5 — Solución final e interpretación física (10 pts)

### 5.1 Solución completa final $u(x,y,t)$
Ensamblando los coeficientes no nulos $A_{2j-1, 2k-1}$ y la dependencia armónica temporal $\cos(\omega_{mn}t)$:

- **En la base de funciones ortonormales $\psi_{mn}$:**
  $$\boxed{u(x,y,t) = \sum_{j=1}^\infty \sum_{k=1}^\infty \frac{8}{\pi^2(2j-1)(2k-1)} \cos\left(c\pi t \sqrt{(2j-1)^2 + (2k-1)^2}\right) \psi_{2j-1, 2k-1}(x,y)}$$

- **En términos directos de las funciones trigonométricas senos:**
  $$\boxed{u(x,y,t) = \frac{16}{\pi^2}\sum_{j=1}^\infty \sum_{k=1}^\infty \frac{\sin\big((2j-1)\pi x\big)\,\sin\big((2k-1)\pi y\big)}{(2j-1)(2k-1)} \cos\left(c\pi t \sqrt{(2j-1)^2 + (2k-1)^2}\right)}$$

---

### 5.2 Modo fundamental y cálculo de su frecuencia exacta $\omega_{11}$
El **modo fundamental** corresponde a los índices más bajos posibles compatibles con la condición de frontera ($m=1, n=1$):
- Autofunción: $\phi_{11}(x,y) = \sin(\pi x)\sin(\pi y)$
- Frecuencia angular exacta:
  $$\boxed{\omega_{11} = c\pi\sqrt{1^2 + 1^2} = \sqrt{2}\,\pi c}$$
- Frecuencia temporal cíclica:
  $$f_{11} = \frac{\omega_{11}}{2\pi} = \frac{\sqrt{2}c}{2} = \frac{c}{\sqrt{2}}$$
- Periodo fundamental de oscilación:
  $$T_{11} = \frac{2\pi}{\omega_{11}} = \frac{\sqrt{2}}{c}$$

> 📊 **Preponderancia energética:** El coeficiente del modo fundamental es $C_{11} = \frac{16}{\pi^2} \approx 1.6211$. Como la energía de cada modo es proporcional a $E_{mn} \propto \omega_{mn}^2 A_{mn}^2$, el modo fundamental contiene más del $81\%$ de la energía total de deformación inicial.

---

### 5.3 Explicación física de la anulación de modos pares
La anulación sistemática de todos los modos con $m$ par o $n$ par obedece a un principio fundamental de conservación: la **simetría de paridad respecto al centro geométrico** de la membrana:
1. El dominio es invariante bajo reflexiones respecto a los ejes axiales centrales $x = 1/2$ e $y = 1/2$.
2. El perfil inicial $u(x,y,0) = 1$ es una constante, por lo que posee **simetría par pura** bajo las transformaciones $\tilde{x} = x - 1/2 \to -\tilde{x}$ y $\tilde{y} = y - 1/2 \to -\tilde{y}$.
3. Las autofunciones espaciales transforman como:
   $$\sin(m\pi x) = \sin\left(m\pi \tilde{x} + \frac{m\pi}{2}\right) = \begin{cases} (-1)^{(m-1)/2} \cos(m\pi \tilde{x}), & m \text{ impar (Función PAR)} \\ (-1)^{m/2} \sin(m\pi \tilde{x}), & m \text{ par (Función IMPAR / ANTISIMÉTRICA)} \end{cases}$$
4. La integral de una función par ($1$) multiplicada por una función impar (modo par) sobre un dominio simétrico es **idénticamente cero**.
5. Físicamente, excitar un modo par requeriría una fuerza o desplazamiento inicial antisimétrico (levantar un lado de la membrana y descender el lado opuesto). Dado que la membrana se deforma homogéneamente hacia arriba en todo punto, la proyección de la fuerza elástica neta sobre cualquier modo par es estrictamente nula.

---

### 5.4 Estado de la membrana en los bordes y en el interior en $t = 0$
En el instante inicial $t = 0$, coexisten dos situaciones físicamente antagónicas:
- **Discontinuidad tipo escalón:** En el interior $0 < x < 1, 0 < y < 1$, la membrana está elevada a altura $u = 1$, pero en los cuatro bordes $x=0,1$ e $y=0,1$, las condiciones de contorno exigen rígidamente $u = 0$.
- **Gradiente y tensión infinitos:** La pendiente espacial $\nabla u$ en las paredes es infinita (un escalón de Heaviside bidimensional). En mecánica del continuo, esto representaría una fuerza de tensión elástica infinita en el contorno.
- **Fenómeno de Gibbs bidimensional:** Como cada término de la serie $\sin(m\pi x)\sin(n\pi y)$ se anula en el borde, la serie de Fourier converge puntualmente a $1$ en el interior abierto y a $0$ en la frontera. Cerca de los bordes, la suma finita produce oscilaciones espaciales rápidas con un sobretiro permanente del $\approx 8.95\%$ a lo largo de las paredes (y hasta $\approx 18\%$ en las esquinas), el cual no desaparece al aumentar el número de modos, sino que se concentra en una banda infinitamente delgada.
- **Dinámica para $t > 0$:** Al soltarse del reposo, las discontinuidades perimetrales actúan como fuentes de ondas de choque de alta frecuencia que viajan hacia el centro a velocidad $c$, generando patrones de interferencia complejos.

---

### 5.5 Líneas nodales del modo $(m,n)$ y mayor frecuencia en modos superiores
1. **Líneas nodales del modo $(m,n)$:**
   Los nodos corresponden a los puntos donde el desplazamiento es nulo en todo tiempo:
   $$\phi_{mn}(x,y) = \sin(m\pi x)\,\sin(n\pi y) = 0$$
   Esto ocurre si $\sin(m\pi x) = 0$ o $\sin(n\pi y) = 0$:
   - En $x$: $x_j = \frac{j}{m}$ para $j = 1, 2, \dots, m-1$. Hay exactamente **$m-1$ líneas nodales verticales interiores**.
   - En $y$: $y_k = \frac{k}{n}$ para $k = 1, 2, \dots, n-1$. Hay exactamente **$n-1$ líneas nodales horizontales interiores**.
   Juntas, forman una cuadrícula ortogonal que divide la membrana en **$m \times n$ celdas rectangulares**, donde celdas adyacentes vibran en estricta contrafase (signos alternados).

2. **¿Por qué los modos superiores oscilan más rápido?**
   - La ecuación de onda establece que la aceleración transversal es proporcional a la **curvatura espacial** (el operador Laplaciano):
     $$u_{tt} = c^2 \nabla^2 u$$
   - Para un modo normal, $\nabla^2 \phi_{mn} = -\lambda_{mn} \phi_{mn} = -(m^2 + n^2)\pi^2 \phi_{mn}$.
   - Modos con $m$ y $n$ grandes poseen una longitud de onda espacial corta ($\lambda_x = 2/m$, $\lambda_y = 2/n$). Esto significa que la membrana se pliega y curva de forma muy pronunciada en distancias cortas.
   - A mayor curvatura, mayor es la deformación relativa entre puntos vecinos y, por tanto, **mayor es la fuerza restauradora elástica neta** por unidad de masa que devuelve a la membrana hacia su posición de equilibrio.
   - Una fuerza restitutiva más potente produce aceleraciones más intensas y, en consecuencia, una **frecuencia de oscilación mucho más elevada**:
     $$\omega_{mn} = c\pi\sqrt{m^2 + n^2} \propto \sqrt{\text{rigidez modal}}$$


```python
# ============================================================
# 🎨 Visualización 2D — FIGURA 3: Reconstrucción 2D de Fourier y Fenómeno de Gibbs
# ============================================================

def serie_fourier_2D(X, Y, N_max):
    """Calcula la suma parcial de Fourier para u(x,y,0) = 1."""
    u_approx = np.zeros_like(X)
    for j in range(1, N_max + 1):
        m = 2 * j - 1
        for k in range(1, N_max + 1):
            n = 2 * k - 1
            coef = 16.0 / (np.pi**2 * m * n)
            u_approx += coef * np.sin(m * np.pi * X) * np.sin(n * np.pi * Y)
    return u_approx

fig = plt.figure(figsize=(16, 10))
fig.patch.set_facecolor('#0d1117')

# Mapas de calor 2D para sumas parciales N_max = 1, 3, 15
terminos = [1, 3, 15]
for idx, N_t in enumerate(terminos):
    ax = fig.add_subplot(2, 3, idx + 1)
    ax.set_facecolor('#0d1117')
    u_t = serie_fourier_2D(X, Y, N_t)
    im = ax.contourf(X, Y, u_t, levels=40, cmap='inferno', vmin=0, vmax=1.2)
    ax.set_title(f'Suma parcial: {N_t}×{N_t} modos impares', fontweight='bold', color='white')
    ax.set_xlabel(r'$x$')
    ax.set_ylabel(r'$y$')
    ax.set_aspect('equal')
    cbar = plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
    cbar.ax.tick_params(colors='white')

# Corte 1D en la línea media y = 0.5
ax_corte1 = fig.add_subplot(2, 2, 3)
ax_corte1.set_facecolor('#0d1117')
ax_corte1.grid(True, alpha=0.15, color='white')

colores = ['#58a6ff', '#f97583', '#7ee787', '#ffa657']
y_mid = np.full_like(x, 0.5)
for N_t, col in zip([1, 3, 7, 25], colores):
    u_eval = serie_fourier_2D(x, y_mid, N_t)
    ax_corte1.plot(x, u_eval, label=f'{N_t}×{N_t} modos', color=col, linewidth=2.0)

ax_corte1.axhline(1.0, color='white', linestyle='--', alpha=0.6, label='Objetivo $u=1$')
ax_corte1.axhline(1.0895, color='#ff7575', linestyle=':', alpha=0.8, label=r'Sobretiro Gibbs $pprox 1.09$')
ax_corte1.set_title(r'Corte transversal en la línea media $y = 0.5$', fontweight='bold')
ax_corte1.set_xlabel(r'$x$')
ax_corte1.set_ylabel(r'$u(x, 0.5, 0)$')
ax_corte1.set_ylim(-0.1, 1.25)
ax_corte1.legend(loc='lower center', fontsize=9, framealpha=0.3, ncol=3)

# Corte 1D en la diagonal x = y
ax_corte2 = fig.add_subplot(2, 2, 4)
ax_corte2.set_facecolor('#0d1117')
ax_corte2.grid(True, alpha=0.15, color='white')

s_diag = np.linspace(0, 1, N)
for N_t, col in zip([1, 3, 7, 25], colores):
    u_diag = np.zeros(N)
    for j in range(1, N_t + 1):
        m = 2 * j - 1
        for k in range(1, N_t + 1):
            n = 2 * k - 1
            coef = 16.0 / (np.pi**2 * m * n)
            u_diag += coef * np.sin(m * np.pi * s_diag) * np.sin(n * np.pi * s_diag)
    ax_corte2.plot(s_diag, u_diag, label=f'{N_t}×{N_t} modos', color=col, linewidth=2.0)

ax_corte2.axhline(1.0, color='white', linestyle='--', alpha=0.6, label='Objetivo $u=1$')
ax_corte2.set_title(r'Corte transversal en la diagonal $x = y$', fontweight='bold')
ax_corte2.set_xlabel(r'$s = x = y$')
ax_corte2.set_ylabel(r'$u(s, s, 0)$')
ax_corte2.set_ylim(-0.1, 1.35)
ax_corte2.legend(loc='lower center', fontsize=9, framealpha=0.3, ncol=3)

fig.suptitle(r'Reconstrucción 2D de Fourier del Estado Inicial $u(x,y,0)=1$ y Fenómeno de Gibbs',
             fontsize=16, color='white', fontweight='bold', y=0.98)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

print("🔍 Observa las oscilaciones de Gibbs cerca de x=0 y x=1.")
print("   Al aumentar el número de modos, el escalón se vuelve más vertical pero el sobretiro (~9%) persiste.")

```

```python
# ============================================================
# 🎨 Visualización 2D — FIGURA 4: Instantáneas Temporales 2D de u(x, y, t)
# ============================================================
c = 1.0
omega_11 = np.sqrt(2) * np.pi * c
T_11 = 2 * np.pi / omega_11  # Periodo del modo fundamental

def u_dinamica(X, Y, t, N_modos=12):
    u = np.zeros_like(X)
    for j in range(1, N_modos + 1):
        m = 2 * j - 1
        for k in range(1, N_modos + 1):
            n = 2 * k - 1
            omega_mn = c * np.pi * np.sqrt(m**2 + n**2)
            coef = 16.0 / (np.pi**2 * m * n)
            u += coef * np.sin(m * np.pi * X) * np.sin(n * np.pi * Y) * np.cos(omega_mn * t)
    return u

instantes = [0.0, T_11/8, T_11/4, 3*T_11/8, T_11/2]
fig, axes = plt.subplots(1, 5, figsize=(18, 4))
fig.patch.set_facecolor('#0d1117')

for ax, t_frac, nombre in zip(axes, instantes, [r'$t = 0$', r'$t = T_{11}/8$', r'$t = T_{11}/4$', r'$t = 3T_{11}/8$', r'$t = T_{11}/2$']):
    ax.set_facecolor('#0d1117')
    u_t = u_dinamica(X, Y, t_frac, N_modos=12)
    im = ax.contourf(X, Y, u_t, levels=35, cmap='magma', vmin=-1.1, vmax=1.1)
    ax.set_title(nombre, color='white', fontweight='bold')
    ax.set_xlabel(r'$x$')
    ax.set_ylabel(r'$y$')
    ax.set_aspect('equal')

cbar_ax = fig.add_axes([0.92, 0.18, 0.015, 0.65])
cbar = fig.colorbar(im, cax=cbar_ax)
cbar.set_label(r'Desplazamiento transversal $u(x,y,t)$', color='white')
cbar.ax.tick_params(colors='white')

fig.suptitle(r'Evolución Temporal 2D de la Membrana Rectangular al Soltarse del Reposo',
             fontsize=15, color='white', fontweight='bold', y=0.98)
plt.tight_layout(rect=[0, 0, 0.91, 0.93])
plt.show()

print("🌊 A t = T_{11}/2, la membrana alcanza su desplazamiento opuesto máximo.")
print("   Las perturbaciones generadas en las fronteras viajan e interfieren en el centro.")

```

```python
# ============================================================
# 🎨 Visualización 2D — FIGURA 5: Retículo Modal y Espectro de Frecuencias
# ============================================================
fig, (ax_grid, ax_stem) = plt.subplots(1, 2, figsize=(14, 5.5))
fig.patch.set_facecolor('#0d1117')

M_max, N_max = 6, 6
m_vals = np.arange(1, M_max + 1)
n_vals = np.arange(1, N_max + 1)
M_grid, N_grid = np.meshgrid(m_vals, n_vals)

A_matrix = np.zeros_like(M_grid, dtype=float)
for i in range(M_max):
    for j in range(N_max):
        m = m_vals[i]
        n = n_vals[j]
        if m % 2 == 1 and n % 2 == 1:
            A_matrix[j, i] = 8.0 / (np.pi**2 * m * n)
        else:
            A_matrix[j, i] = 0.0

# Retículo 2D de modos (m, n)
for i in range(M_max):
    for j in range(N_max):
        m = m_vals[i]
        n = n_vals[j]
        amp = A_matrix[j, i]
        omega_rel = np.sqrt(m**2 + n**2) / np.sqrt(2)
        if amp > 0:
            ax_grid.scatter(m, n, s=amp*1200, color='#58a6ff', alpha=0.85, edgecolors='white', linewidth=1.5)
            ax_grid.text(m, n, f'{omega_rel:.2f} ' + r'$\omega_{11}$' + f'\n({amp:.2f})',
                         ha='center', va='center', color='white', fontsize=8, fontweight='bold')
        else:
            ax_grid.scatter(m, n, s=40, color='#f97583', alpha=0.3, marker='x')

ax_grid.set_title(r'Retículo Cuantizado $(m,n)$ y Factor $\omega_{mn}/\omega_{11}$' + '\n' + 
                  '(Burbujas azules = Modos activos; Cruces rojas = Modos nulos)', fontweight='bold')
ax_grid.set_xlabel(r'Modo $m$ en dirección $x$')
ax_grid.set_ylabel(r'Modo $n$ en dirección $y$')
ax_grid.set_xticks(m_vals)
ax_grid.set_yticks(n_vals)
ax_grid.grid(True, alpha=0.15, color='white')

# Stem plot de amplitudes relativas en orden de frecuencia creciente
modos_activos = []
for m in range(1, 10, 2):
    for n in range(1, 10, 2):
        amp = 16.0 / (np.pi**2 * m * n)
        freq = np.sqrt(m**2 + n**2) / np.sqrt(2)
        modos_activos.append((freq, amp, f'({m},{n})'))

modos_activos.sort(key=lambda item: item[0])
freqs = [item[0] for item in modos_activos]
amps = [item[1] for item in modos_activos]
labels = [item[2] for item in modos_activos]

ax_stem.stem(freqs[:12], amps[:12], linefmt='#7ee787', markerfmt='o', basefmt='white')
for f, a, lbl in zip(freqs[:7], amps[:7], labels[:7]):
    ax_stem.annotate(lbl, (f, a), textcoords="offset points", xytext=(0, 7),
                     ha='center', fontsize=9, color='white', fontweight='bold')

ax_stem.set_title(r'Decaimiento Espectral $C_{mn} = \frac{16}{\pi^2 m n}$ vs Frecuencia', fontweight='bold')
ax_stem.set_xlabel(r'Frecuencia relativa $\omega_{mn} / \omega_{11}$')
ax_stem.set_ylabel(r'Amplitud del modo')
ax_stem.set_ylim(0, 1.8)
ax_stem.grid(True, alpha=0.15, color='white')

plt.tight_layout()
plt.show()

print("💡 El modo fundamental (1,1) domina la dinámica con diferencia.")
print("   Las amplitudes decaen cuadráticamente como 1/(m·n) para altas frecuencias.")

```

```python
# ============================================================
# 🌐 Superficie 3D interactiva con k3d
# ============================================================
# ¡Arrastra con el mouse para rotar, haz zoom y explora en 3D!
# ============================================================

import k3d
from k3d.colormaps import matplotlib_color_maps

# Parámetros del modo a explorar estáticamente
m_explorar = 3
n_explorar = 1

# Malla para k3d
N_k3d = 150
x_k = np.linspace(0.0, 1.0, N_k3d)
y_k = np.linspace(0.0, 1.0, N_k3d)
X_k, Y_k = np.meshgrid(x_k, y_k)

# Modo normal estático en t = 0
Z_estatico = np.sin(m_explorar * np.pi * X_k) * np.sin(n_explorar * np.pi * Y_k)
# Garantizar bordes fijos a cero
Z_estatico[0, :] = 0; Z_estatico[-1, :] = 0; Z_estatico[:, 0] = 0; Z_estatico[:, -1] = 0

Z_estatico_32 = Z_estatico.astype(np.float32)

plot_3d = k3d.plot(
    grid_auto_fit=True,
    camera_auto_fit=True,
    height=550
)

superficie_3d = k3d.surface(
    Z_estatico_32,
    attribute=Z_estatico_32,
    xmin=0.0, xmax=1.0,
    ymin=0.0, ymax=1.0,
    color_map=matplotlib_color_maps.Coolwarm,
    color_range=[-1.0, 1.0],
    flat_shading=False
)

plot_3d += superficie_3d
plot_3d.display()

print(f"🔲 Modo Normal ({m_explorar}, {n_explorar}) renderizado en 3D WebGL.")
print("   → Click izquierdo + arrastrar: Rotar cámara.")
print("   → Scroll del mouse: Zoom.")
print("   → Click derecho + arrastrar: Trasladar.")

```

### 🎬 Animación en Tiempo Real: ¡La Membrana Rectangular Vibra en Vivo!

En la siguiente celda presentamos el **laboratorio interactivo de vibración en tiempo real**. El reproductor cuenta con:
- Botón **▶ Play / Pausa**: Para iniciar o detener la vibración armónica continua.
- Slider de **Tiempo**: Para avanzar cuadro por cuadro y congelar cualquier fase de la onda.
- Selector de **Simulación**:
  1. **Solución Completa Examen $u(x,y,t)$:** Observa cómo la membrana, soltada desde el reposo con elevación plana $u_0 = 1$ y bordes fijos $u=0$, colapsa y genera frentes de onda de choque que viajan e interfieren en el centro.
  2. **Modo Normal $(m,n)$:** Selecciona números enteros $m$ y $n$ con los sliders para aislar cualquier armónico puro de la membrana.


```python
# ============================================================
# 🎬 Laboratorio Interactivo con k3d + ipywidgets
# ============================================================
# ¡Presiona Play para ver la membrana rectangular vibrar en 3D!
# ============================================================

import ipywidgets as widgets
from IPython.display import display
import k3d
from k3d.colormaps import matplotlib_color_maps

# Malla y parámetros físicos
N_anim = 100
x_anim = np.linspace(0.0, 1.0, N_anim)
y_anim = np.linspace(0.0, 1.0, N_anim)
X_anim, Y_anim = np.meshgrid(x_anim, y_anim)

c_onda = 1.0
omega_11 = np.sqrt(2) * np.pi * c_onda
T_11 = 2 * np.pi / omega_11  # Periodo del modo fundamental
N_frames = 100

def calcular_Z_anim(tipo, m, n, t, n_fourier=10):
    if tipo == 'Modo Normal (m,n)':
        om = c_onda * np.pi * np.sqrt(m**2 + n**2)
        Z = np.sin(m * np.pi * X_anim) * np.sin(n * np.pi * Y_anim) * np.cos(om * t)
    else: # Solución Completa del Examen u(x,y,t)
        Z = np.zeros_like(X_anim)
        for j in range(1, n_fourier + 1):
            mj = 2 * j - 1
            for k in range(1, n_fourier + 1):
                nk = 2 * k - 1
                om = c_onda * np.pi * np.sqrt(mj**2 + nk**2)
                coef = 16.0 / (np.pi**2 * mj * nk)
                Z += coef * np.sin(mj * np.pi * X_anim) * np.sin(nk * np.pi * Y_anim) * np.cos(om * t)
                
    # Fronteras rígidas en cero
    Z[0, :] = 0.0; Z[-1, :] = 0.0; Z[:, 0] = 0.0; Z[:, -1] = 0.0
    return Z.astype(np.float32)

# Crear plot de k3d
plot_anim = k3d.plot(
    grid_auto_fit=False,
    camera_auto_fit=False,
    height=600
)

# Superficie inicial en t = 0
Z_inicial = calcular_Z_anim('Solución Completa Examen u(x,y,t)', 1, 1, 0.0, n_fourier=10)

superficie_anim = k3d.surface(
    Z_inicial,
    attribute=Z_inicial,
    xmin=0.0, xmax=1.0,
    ymin=0.0, ymax=1.0,
    color_map=matplotlib_color_maps.Coolwarm,
    color_range=[-1.15, 1.15],
    flat_shading=False
)
plot_anim += superficie_anim

# ──────────────────────────────────────────────────────────
# Widgets Interactivos
# ──────────────────────────────────────────────────────────
play = widgets.Play(
    value=0, min=0, max=N_frames - 1, step=1,
    interval=70, description="Play"
)
slider_t = widgets.IntSlider(
    min=0, max=N_frames - 1, step=1,
    description='Tiempo:',
    layout=widgets.Layout(width='360px')
)
widgets.jslink((play, 'value'), (slider_t, 'value'))

dropdown_tipo = widgets.Dropdown(
    options=['Solución Completa Examen u(x,y,t)', 'Modo Normal (m,n)'],
    value='Solución Completa Examen u(x,y,t)',
    description='Simulación:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='340px')
)

slider_m = widgets.IntSlider(value=1, min=1, max=6, step=1, description='m (en x):', layout=widgets.Layout(width='200px'))
slider_n = widgets.IntSlider(value=1, min=1, max=6, step=1, description='n (en y):', layout=widgets.Layout(width='200px'))
slider_fourier = widgets.IntSlider(
    value=10, min=1, max=25, step=1,
    description='Términos Fourier:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='270px')
)

caja_modos = widgets.HBox([slider_m, slider_n])
caja_fourier = widgets.HBox([slider_fourier])

def actualizar_frame(change=None):
    frame = slider_t.value
    t_val = (frame / N_frames) * T_11
    tipo = dropdown_tipo.value
    m_val = slider_m.value
    n_val = slider_n.value
    n_f = slider_fourier.value
    
    Z_nuevo = calcular_Z_anim(tipo, m_val, n_val, t_val, n_f)
    superficie_anim.heights = Z_nuevo
    superficie_anim.attribute = Z_nuevo

def alternar_interfaz(change=None):
    if dropdown_tipo.value == 'Modo Normal (m,n)':
        caja_modos.layout.display = 'flex'
        caja_fourier.layout.display = 'none'
    else:
        caja_modos.layout.display = 'none'
        caja_fourier.layout.display = 'flex'
    actualizar_frame()

slider_t.observe(actualizar_frame, names='value')
dropdown_tipo.observe(alternar_interfaz, names='value')
slider_m.observe(actualizar_frame, names='value')
slider_n.observe(actualizar_frame, names='value')
slider_fourier.observe(actualizar_frame, names='value')

caja_modos.layout.display = 'none'
caja_fourier.layout.display = 'flex'

controles_sup = widgets.HBox([dropdown_tipo, caja_fourier, caja_modos])
controles_inf = widgets.HBox([play, slider_t])

display(widgets.VBox([controles_sup, controles_inf]))
plot_anim.display()

print("🎬 ¡Presiona ▶ Play para iniciar la vibración 3D en tiempo real!")
print("   Cambia el selector para alternar entre el examen completo y modos individuales.")

```
