Derivaremos paso a paso la ecuación diferencial de Bessel, exploraremos sus soluciones gráficamente y visualizaremos los **modos normales de vibración** en 3D.

## 1. Planteamiento del problema

Consideremos una membrana circular delgada de radio $a$, perfectamente flexible y uniforme, sujeta en todo su borde circular. Cuando se perturba y se suelta, vibra transversalmente.

La amplitud transversal $U(r, \phi, t)$ satisface la **ecuación de onda bidimensional**:

$$\nabla^2 U - \frac{1}{v^2}\frac{\partial^2 U}{\partial t^2} = 0$$

donde $v$ es la velocidad de propagación de las ondas en la membrana (determinada por la tensión y la densidad superficial).

### El Laplaciano en coordenadas polares

Dado que la membrana es un **disco**, la simetría del problema nos pide usar coordenadas polares $(r, \phi)$. En estas coordenadas, el operador Laplaciano toma la forma:

$$\nabla^2 U = \frac{\partial^2 U}{\partial r^2} + \frac{1}{r}\frac{\partial U}{\partial r} + \frac{1}{r^2}\frac{\partial^2 U}{\partial \phi^2}$$

Sustituyendo en la ecuación de onda:

$$\frac{\partial^2 U}{\partial r^2} + \frac{1}{r}\frac{\partial U}{\partial r} + \frac{1}{r^2}\frac{\partial^2 U}{\partial \phi^2} = \frac{1}{v^2}\frac{\partial^2 U}{\partial t^2}$$

## 2. Separación de variables

Buscamos soluciones de la forma **separable**, es decir, proponemos que $U$ se puede escribir como un producto de funciones, cada una dependiente de una sola variable:

$$U(r, \phi, t) = R(r)\,\Phi(\phi)\,T(t)$$

Sustituimos esta propuesta en la ecuación de onda. Usando la notación de primas para derivadas respecto a la variable correspondiente:

$$R''\,\Phi\,T + \frac{1}{r}R'\,\Phi\,T + \frac{1}{r^2}R\,\Phi''\,T = \frac{1}{v^2}R\,\Phi\,T''$$

Dividimos todo entre $R\,\Phi\,T$:

$$\frac{R''}{R} + \frac{1}{r}\frac{R'}{R} + \frac{1}{r^2}\frac{\Phi''}{\Phi} = \frac{1}{v^2}\frac{T''}{T}$$

El lado izquierdo depende solo de $r$ y $\phi$; el lado derecho solo de $t$. Como son iguales para **todo** valor de las tres variables, ambos lados deben ser iguales a una misma **constante de separación** $-k^2$ (elegimos el signo negativo para obtener soluciones oscilatorias):

$$\frac{R''}{R} + \frac{1}{r}\frac{R'}{R} + \frac{1}{r^2}\frac{\Phi''}{\Phi} = \frac{1}{v^2}\frac{T''}{T} = -k^2$$

### 2.1 Ecuación temporal

Del lado derecho de la separación obtenemos:

$$\frac{1}{v^2}\frac{T''}{T} = -k^2 \quad \Longrightarrow \quad T'' + v^2 k^2\,T = 0$$

Definiendo la **frecuencia angular** $\omega = vk$, la ecuación se convierte en:

$$T'' + \omega^2 T = 0$$

cuya solución general es una **oscilación armónica**:

$$\boxed{T(t) = A\cos(\omega t) + B\sin(\omega t)}$$

Las constantes $A$ y $B$ se determinan por las condiciones iniciales (posición y velocidad de la membrana en $t=0$).

### 2.2 Ecuación angular

Tomamos la parte espacial y multiplicamos por $r^2$:

$$r^2\frac{R''}{R} + r\frac{R'}{R} + k^2 r^2 + \frac{\Phi''}{\Phi} = 0$$

Reagrupamos separando las partes que dependen de $r$ y de $\phi$:

$$r^2\frac{R''}{R} + r\frac{R'}{R} + k^2 r^2 = -\frac{\Phi''}{\Phi} = m^2$$

donde $m^2$ es una nueva constante de separación. Esto nos da la **ecuación angular**:

$$\Phi'' + m^2\Phi = 0$$

cuya solución general es:

$$\Phi(\phi) = C\cos(m\phi) + D\sin(m\phi)$$

**Condición de periodicidad:** La membrana es un objeto físico continuo, así que la función $\Phi$ debe ser periódica con periodo $2\pi$:

$$\Phi(\phi) = \Phi(\phi + 2\pi)$$

Esto exige que $m$ sea un **número entero no negativo**: $m = 0, 1, 2, 3, \ldots$

> 💡 **Interpretación física:** El número $m$ cuenta las **líneas nodales diametrales** del modo de vibración.

### 2.3 Ecuación radial: la ecuación de Bessel

Usando la parte radial de la separación:

$$r^2\frac{R''}{R} + r\frac{R'}{R} + k^2 r^2 = m^2$$

Reagrupamos en forma estándar:

$$r^2 R'' + r R' + (k^2 r^2 - m^2)\,R = 0$$

> 🎉 **¡Esta es la ecuación diferencial de Bessel de orden $m$!**

Es una de las ecuaciones más importantes de la física matemática. Aparece siempre que se usa separación de variables en coordenadas cilíndricas o polares.

## 3. Funciones de Bessel: $J_m$ y $Y_m$

La solución general de la ecuación de Bessel de orden $m$ es una combinación lineal de dos funciones linealmente independientes:

$$R(r) = E\,J_m(kr) + F\,Y_m(kr)$$

donde:
- $J_m(x)$ es la **función de Bessel de primera especie** de orden $m$
- $Y_m(x)$ es la **función de Bessel de segunda especie** de orden $m$

### ¿Por qué descartamos $Y_m$?

La función $Y_m(x)$ tiene una propiedad crucial: **diverge** (se va a $-\infty$) cuando $x \to 0$:

$$Y_m(x) \xrightarrow{x \to 0} -\infty$$

Pero el centro de nuestra membrana ($r = 0$) es un punto físico perfectamente válido donde la amplitud debe ser finita. Por lo tanto, debemos exigir $F = 0$, y la solución radial queda:

$$\boxed{R(r) = E\,J_m(kr)}$$

## 4. Condiciones de frontera y modos normales

### La membrana está sujeta en el borde

La membrana está firmemente sujeta en su borde circular de radio $a$. Esto significa que el desplazamiento es cero en $r = a$ para todo ángulo $\phi$ y todo tiempo $t$:

$$U(a, \phi, t) = 0$$

Sustituyendo nuestra solución separable:

$$R(a)\,\Phi(\phi)\,T(t) = 0 \quad \forall\, \phi, t$$

Como no queremos la solución trivial ($\Phi = 0$ o $T = 0$), necesitamos:

$$R(a) = 0 \quad \Longrightarrow \quad J_m(ka) = 0$$

### Los ceros de Bessel determinan las frecuencias

Definimos $\alpha_{mn}$ como el **$n$-ésimo cero** de la función de Bessel $J_m$:

$$J_m(\alpha_{mn}) = 0 \qquad n = 1, 2, 3, \ldots$$

La condición $J_m(ka) = 0$ entonces implica:

$$k_{mn} = \frac{\alpha_{mn}}{a}$$

y las frecuencias de vibración permitidas son:

$$\omega_{mn} = v\,k_{mn} = \frac{v\,\alpha_{mn}}{a}$$

> ⚠️ **Nota importante:** A diferencia de la cuerda vibrante, las frecuencias $\omega_{mn}$ del tambor **no son** múltiplos enteros de una frecuencia fundamental. Por eso el sonido de un tambor no tiene una altura musical bien definida.

```python
# ============================================================
# 📋 Tabla de ceros de las funciones de Bessel J_m(x)
# ============================================================
# Los ceros α_{m,n} son los valores de x donde J_m(x) = 0.
# Determinan las frecuencias de vibración de la membrana.
# ============================================================

from scipy.special import jn_zeros   # Calcula ceros de J_m

# --- Encabezado de la tabla ---
print("=" * 62)
print("   Ceros de las funciones de Bessel: α_{m,n}")
print("   Definidos por J_m(α_{m,n}) = 0")
print("=" * 62)

# Encabezado de columnas
encabezado = "  m \\ n  |"
for n in range(1, 6):
    encabezado += f"   n={n}    |"
print(encabezado)
print("-" * 62)

# --- Calculamos y mostramos los ceros para m = 0, 1, 2, 3 ---
for m in range(4):
    # jn_zeros(m, 5) devuelve un array con los primeros 5 ceros de J_m
    ceros = jn_zeros(m, 5)

    fila = f"  m={m}   |"
    for cero in ceros:
        fila += f" {cero:8.4f} |"
    print(fila)

print("=" * 62)
print()
print("💡 Los ceros NO están espaciados uniformemente.")
print("   Esto explica por qué el tambor no tiene armónicos simples.")
```

### Solución completa: el modo normal $U_{mn}$

Juntando todas las piezas, el **modo normal de vibración** $(m, n)$ es:

$$\boxed{U_{mn}(r, \phi, t) = J_m\!\left(\frac{\alpha_{mn}}{a}\,r\right)\left[C\cos(m\phi) + D\sin(m\phi)\right]\cos(\omega_{mn}\,t)}$$

donde:
- $m = 0, 1, 2, \ldots$ → número de **líneas nodales diametrales** (cortes radiales donde la membrana no se mueve)
- $n = 1, 2, 3, \ldots$ → número de **círculos nodales** (anillos donde la membrana no se mueve, incluyendo el borde)
- $\alpha_{mn}$ → $n$-ésimo cero de $J_m$
- $\omega_{mn} = v\,\alpha_{mn}/a$ → frecuencia angular del modo

### Nomenclatura de los modos

| Modo $(m, n)$ | Líneas nodales | Descripción |
|:---:|:---:|:---|
| $(0, 1)$ | Solo el borde | **Modo fundamental** — toda la membrana sube y baja |
| $(1, 1)$ | 1 diámetro + borde | La membrana se divide en 2 mitades que vibran en contrafase |
| $(2, 1)$ | 2 diámetros + borde | 4 cuadrantes vibrando alternadamente |
| $(0, 2)$ | 1 círculo + borde | Un anillo central y otro exterior, en contrafase |

## 5. Visualización 3D de los modos normales

Ahora viene la parte más divertida: **ver** cómo vibra la membrana.

Representaremos la membrana como una superficie 3D donde:
- Las coordenadas $(x, y)$ son la posición en el plano de la membrana
- La coordenada $z$ es el **desplazamiento vertical** $U_{mn}(r, \phi, t)$

Primero crearemos una visualización **estática** con `matplotlib` (4 modos en $t=0$), y después una visualización **interactiva y animada** con `k3d`.

```python
# ============================================================
# 🧮 Funciones auxiliares para construir la malla y calcular
#    los modos normales de vibración de la membrana
# ============================================================

import numpy as np
from scipy.special import jv, jn_zeros


def crear_malla_disco(N=100, radio=1.0):
    """
    Crea una malla rectangular que cubre el disco de radio dado.

    Parámetros
    ----------
    N : int
        Número de puntos por lado de la malla cuadrada.
        Más puntos = más resolución, pero más lento.
    radio : float
        Radio de la membrana circular.

    Retorna
    -------
    x_grid, y_grid : arrays 2D de forma (N, N)
        Coordenadas cartesianas de cada punto de la malla.
    r_grid, phi_grid : arrays 2D de forma (N, N)
        Coordenadas polares (r, phi) de cada punto.
    mascara : array 2D booleano de forma (N, N)
        True en los puntos que están DENTRO del disco (r <= radio).
    """
    # Paso 1: Crear un arreglo de puntos de -radio a +radio
    eje = np.linspace(-radio, radio, N)

    # Paso 2: Crear la malla 2D (todos los pares (x, y))
    x_grid, y_grid = np.meshgrid(eje, eje)

    # Paso 3: Convertir a coordenadas polares
    r_grid = np.sqrt(x_grid**2 + y_grid**2)      # r = sqrt(x^2 + y^2)
    phi_grid = np.arctan2(y_grid, x_grid)          # phi = arctan(y/x)

    # Paso 4: Crear la máscara del disco
    mascara = r_grid <= radio   # True si el punto está dentro del disco

    return x_grid, y_grid, r_grid, phi_grid, mascara


def modo_normal(m, n, r, phi, t, radio=1.0, amplitud=0.3):
    """
    Calcula el desplazamiento vertical del modo normal (m, n).

    La fórmula es:
        U_{m,n} = A * J_m(k_{m,n} * r) * cos(m*phi) * cos(omega_{m,n} * t)

    Parámetros
    ----------
    m : int
        Número angular (líneas nodales diametrales). m = 0, 1, 2, ...
    n : int
        Número radial (círculos nodales). n = 1, 2, 3, ...
    r : array
        Coordenada radial de cada punto.
    phi : array
        Coordenada angular de cada punto.
    t : float
        Instante de tiempo.
    radio : float
        Radio de la membrana (a).
    amplitud : float
        Amplitud máxima del desplazamiento.

    Retorna
    -------
    Z : array
        Desplazamiento vertical U en cada punto.
    """
    # Paso 1: Obtener el n-ésimo cero de J_m
    # jn_zeros(m, n) devuelve los primeros n ceros de J_m
    alpha_mn = jn_zeros(m, n)[-1]    # Tomamos el último (= el n-ésimo)

    # Paso 2: Calcular el número de onda k = alpha_{m,n} / a
    k = alpha_mn / radio

    # Paso 3: Calcular la frecuencia angular omega = v * k
    # Normalizamos v = 1, entonces omega = k = alpha_{m,n} / radio
    omega = alpha_mn / radio

    # Paso 4: Evaluar la fórmula del modo normal
    Z = amplitud * jv(m, k * r) * np.cos(m * phi) * np.cos(omega * t)

    return Z


print("✅ Funciones auxiliares definidas correctamente")
print("   → crear_malla_disco(N, radio)")
print("   → modo_normal(m, n, r, phi, t)")
```

```python
# ============================================================
# 🖼️ Visualización estática de 4 modos normales (matplotlib)
# ============================================================
# Esta celda genera una figura con los 4 modos más comunes
# del tambor, congelados en el instante t = 0.
# ============================================================

from mpl_toolkits.mplot3d import Axes3D   # Soporte para gráficas 3D

# --- Crear la malla del disco ---
N = 150                          # Resolución: 150 x 150 puntos
x, y, r, phi, mascara = crear_malla_disco(N=N, radio=1.0)

# --- Definir los 4 modos a graficar ---
# Cada tupla es (m, n): m = líneas nodales, n = círculos nodales
# ╔══════════════════════════════════════════════════════════════╗
# ║  👇 CAMBIA ESTOS VALORES para explorar otros modos          ║
# ╚══════════════════════════════════════════════════════════════╝
modos = [(0, 1), (1, 1), (2, 1), (0, 2)]

# Títulos descriptivos para cada modo
titulos = [
    r'$(0,1)$: Fundamental',
    r'$(1,1)$: 1 línea nodal',
    r'$(2,1)$: 2 líneas nodales',
    r'$(0,2)$: 1 círculo nodal',
]

# --- Crear la figura con 4 subplots ---
fig = plt.figure(figsize=(16, 13))
fig.patch.set_facecolor('#0d1117')

for idx, ((m, n), titulo) in enumerate(zip(modos, titulos)):
    # Crear subplot 3D (2 filas x 2 columnas)
    ax = fig.add_subplot(2, 2, idx + 1, projection='3d')
    ax.set_facecolor('#0d1117')

    # Calcular el desplazamiento Z en t = 0
    Z = modo_normal(m, n, r, phi, t=0)

    # Poner Z = NaN fuera del disco (no se grafica)
    Z[~mascara] = np.nan

    # Graficar la superficie 3D
    surf = ax.plot_surface(
        x, y, Z,
        cmap='coolwarm',       # Mapa de color: azul (neg.) ↔ rojo (pos.)
        edgecolor='none',      # Sin bordes entre celdas
        alpha=0.95,            # Ligera transparencia
        rstride=2, cstride=2,  # Paso de muestreo (2 = cada 2 puntos)
        antialiased=True
    )

    # Configurar el aspecto del subplot
    ax.set_title(f'Modo {titulo}', fontsize=13, color='white',
                 pad=10, fontweight='bold')
    ax.set_xlabel(r'$x$', color='white')
    ax.set_ylabel(r'$y$', color='white')
    ax.set_zlabel(r'$U$', color='white')
    ax.set_zlim(-0.35, 0.35)

    # Ángulo de vista elevado para apreciar el patrón
    ax.view_init(elev=25, azim=-60)

    # Color de los ejes
    ax.tick_params(colors='white')
    ax.xaxis.pane.fill = False
    ax.yaxis.pane.fill = False
    ax.zaxis.pane.fill = False

# Título general
fig.suptitle(
    r'Modos normales de vibración $U_{m,n}(r, \phi, t\!=\!0)$',
    fontsize=17, color='white', fontweight='bold', y=0.95
)

plt.tight_layout(rect=[0, 0, 1, 0.93])
plt.show()

print("💡 Observa cómo cada modo tiene un patrón nodal diferente.")
print("   Los nodos son las regiones donde la membrana NO se mueve (Z = 0).")
```

```python
# ============================================================
# 🌐 Visualización 3D interactiva con k3d
# ============================================================
# k3d permite ROTAR, hacer ZOOM y explorar la superficie
# directamente en el notebook. ¡Pruébalo!
# ============================================================

import k3d
from k3d.colormaps import matplotlib_color_maps

# ╔══════════════════════════════════════════════════════════════╗
# ║  👇 CAMBIA ESTOS VALORES para ver otros modos               ║
# ╚══════════════════════════════════════════════════════════════╝
m_modo = 2    # Número angular (m = 0, 1, 2, 3, ...)
n_modo = 1    # Número radial  (n = 1, 2, 3, ...)
# ══════════════════════════════════════════════════════════════════

# --- Crear la malla (alta resolución para k3d) ---
N_k3d = 200                     # 200 x 200 puntos
x_k, y_k, r_k, phi_k, mask_k = crear_malla_disco(N=N_k3d, radio=1.0)

# --- Calcular el desplazamiento Z en t = 0 ---
Z_k = modo_normal(m_modo, n_modo, r_k, phi_k, t=0)
Z_k[~mask_k] = 0               # Z = 0 fuera del disco

# --- Convertir a float32 (requerido por WebGL / k3d) ---
Z_k32 = Z_k.astype(np.float32)

# --- Crear el plot k3d ---
plot = k3d.plot(
    grid_auto_fit=True,         # Ajustar automáticamente la cuadrícula
    camera_auto_fit=True,       # Ajustar cámara al objeto
    height=600                  # Altura del widget en píxeles
)

# --- Crear la superficie ---
superficie = k3d.surface(
    Z_k32,                      # Array 2D de alturas (float32)
    xmin=-1.0, xmax=1.0,        # Rango en x
    ymin=-1.0, ymax=1.0,        # Rango en y
    color_map=matplotlib_color_maps.Coolwarm,   # Azul ↔ Rojo
    color_range=[-0.3, 0.3],    # Rango de colores (centrado en 0)
    flat_shading=False          # Sombreado suave (interpolado)
)

# Agregar la superficie al plot y mostrarlo
plot += superficie
plot.display()

print(f"🥁 Modo (m={m_modo}, n={n_modo})")
print("   → Arrastra con el mouse para rotar")
print("   → Scroll para hacer zoom")
print("   → Click derecho + arrastrar para desplazar")
```

```python
# ============================================================
# 🎬 Animación interactiva: ¡la membrana vibra!
# ============================================================
# Usa el botón ▶ Play para ver la vibración en tiempo real.
# Puedes cambiar el modo (m, n) con los menús desplegables.
# ============================================================

import ipywidgets as widgets       # Widgets interactivos de Jupyter
from IPython.display import display

# --- Crear nuevo plot k3d para la animación ---
plot_anim = k3d.plot(
    grid_auto_fit=False,
    camera_auto_fit=False,
    height=600
)

# --- Parámetros iniciales ---
m_inicial = 2      # Modo angular inicial
n_inicial = 1      # Modo radial inicial
N_frames = 100     # Número total de cuadros en la animación

# --- Superficie inicial (t = 0) ---
Z_init = modo_normal(m_inicial, n_inicial, r_k, phi_k, t=0)
Z_init[~mask_k] = 0

superficie_anim = k3d.surface(
    Z_init.astype(np.float32),
    xmin=-1.0, xmax=1.0,
    ymin=-1.0, ymax=1.0,
    color_map=matplotlib_color_maps.Coolwarm,
    color_range=[-0.3, 0.3],
    flat_shading=False
)
plot_anim += superficie_anim

# ──────────────────────────────────────────────────────────
# Widgets de control
# ──────────────────────────────────────────────────────────

# Botón de play/pausa automático
play = widgets.Play(
    value=0,
    min=0,
    max=N_frames - 1,
    step=1,
    interval=80,           # Milisegundos entre cuadros (~12.5 fps)
    description="Play"
)

# Slider manual de tiempo
slider_t = widgets.IntSlider(
    min=0,
    max=N_frames - 1,
    step=1,
    description='Tiempo:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='400px')
)

# Vincular play ↔ slider (se mueven juntos automáticamente)
widgets.jslink((play, 'value'), (slider_t, 'value'))

# Menú desplegable para elegir m (número angular)
dropdown_m = widgets.Dropdown(
    options=[0, 1, 2, 3, 4, 5],
    value=m_inicial,
    description='m (angular):',
    style={'description_width': 'initial'}
)

# Menú desplegable para elegir n (número radial)
dropdown_n = widgets.Dropdown(
    options=[1, 2, 3, 4],
    value=n_inicial,
    description='n (radial):',
    style={'description_width': 'initial'}
)

# ──────────────────────────────────────────────────────────
# Función que actualiza la superficie en cada cuadro
# ──────────────────────────────────────────────────────────

def actualizar_superficie(change):
    """
    Se ejecuta automáticamente cada vez que cambia el slider
    de tiempo o los menús desplegables de m y n.
    """
    # Leer el cuadro actual del slider
    frame = slider_t.value

    # Leer los valores actuales de m y n
    m_actual = dropdown_m.value
    n_actual = dropdown_n.value

    # Obtener el cero de Bessel para el modo actual
    alpha_mn = jn_zeros(m_actual, n_actual)[-1]

    # Calcular el tiempo real para que un ciclo completo de
    # animación (frame 0 → N_frames) corresponda exactamente
    # a un periodo de oscilación T = 2π/ω.
    #
    # Como ω = α_{m,n}/a (con a=1), entonces:
    #   t = (2π · frame) / (N_frames · α_{m,n})
    # y ω·t = 2π·frame/N_frames → un ciclo completo ✓
    t = (2 * np.pi * frame) / (N_frames * alpha_mn)

    # Recalcular el desplazamiento Z
    Z_new = modo_normal(m_actual, n_actual, r_k, phi_k, t=t)
    Z_new[~mask_k] = 0

    # Actualizar la superficie en k3d (¡sin recrear el plot!)
    superficie_anim.heights = Z_new.astype(np.float32)

# Conectar la función de actualización a los widgets
slider_t.observe(actualizar_superficie, names='value')
dropdown_m.observe(actualizar_superficie, names='value')
dropdown_n.observe(actualizar_superficie, names='value')

# ──────────────────────────────────────────────────────────
# Mostrar controles y plot
# ──────────────────────────────────────────────────────────
selectores = widgets.HBox([dropdown_m, dropdown_n])
controles = widgets.HBox([play, slider_t])

display(widgets.VBox([selectores, controles]))
plot_anim.display()

print("🎬 ¡Presiona ▶ Play para ver la membrana vibrar!")
print("   Cambia m y n en los menús para explorar otros modos.")
```
